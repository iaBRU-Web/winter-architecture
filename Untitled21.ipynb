```python
#!/usr/bin/env python
"""
BarameterLM - one file: config, model, barameter memory, training, ingestion, search, refresh, growth, CLI.

  python barameter.py selftest                                   # CPU self-test (no network)
  python barameter.py train --dataset yahma/alpaca-cleaned --minutes 10 --hub_repo brucoder/BarameterLM
  python barameter.py train --resume --hub_repo brucoder/BarameterLM --minutes 10
  python barameter.py ingest  --ckpt brucoder/BarameterLM --dataset databricks/databricks-dolly-15k --tag dolly
  python barameter.py ask     --ckpt brucoder/BarameterLM "What is the capital of France?" [--deep]
  python barameter.py chat    --ckpt brucoder/BarameterLM
  python barameter.py search  --ckpt brucoder/BarameterLM "query"       # deep GPU search over every stored chunk
  python barameter.py refresh --ckpt brucoder/BarameterLM --steps 20    # find forgotten chunks, rewrite + short retrain
  python barameter.py reindex --ckpt brucoder/BarameterLM               # re-encode all keys with the current writer
  python barameter.py eval    --ckpt brucoder/BarameterLM --dataset yahma/alpaca-cleaned
  python barameter.py list | inspect <slot> | delete <tag>  --ckpt ...
  python barameter.py grow    --ckpt ... --add_layers 12 --d_model 1280 --out grown [--push brucoder/BarameterLM-300M]

Memory design
  hot tier  : BarameterStore  - fixed slots, L2-normalised 128-d keys + 4x8-bit product-quantised values (4 bytes/slot),
              evicts least-used when full, merges near-duplicate keys, every entry tagged with a dataset.
  cold tier : ChunkIndex      - EVERY chunk ever seen (key + tokens), brute-force GPU search, used by the deep path.
  Search is chunked over slots and arrays are numpy-backed (optionally np.memmap, --mmap), so the store can grow
  from 65k slots in VRAM to millions of disk-mapped slots without touching the backbone.
Auth: never hard-coded. Uses HF_TOKEN env / Colab secret HF_TOKEN / cached login, else an interactive login().
"""
import os, sys, json, math, time, copy, queue, shutil, hashlib, argparse, random, threading, tempfile
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.checkpoint import checkpoint
from transformers import PretrainedConfig, PreTrainedModel, AutoConfig, AutoModelForCausalLM
from transformers.modeling_outputs import CausalLMOutput

MODULE_NAME = os.path.splitext(os.path.basename(globals().get("__file__", "barameter.py")))[0]
SPECIALS = ["<pad>", "<eos>", "<|user|>", "<|assistant|>"]


# ----------------------------------------------------------------------------------------------
# Config
# ----------------------------------------------------------------------------------------------
class BarameterConfig(PretrainedConfig):
    model_type = "barameter"

    def __init__(self, vocab_size=16384, d_model=640, n_heads=10, n_kv_heads=2, head_dim=64, ffn_hidden=1728,
                 block_types=None, read_after=(4, 8), write_from=6, rec_chunk=64, write_chunk=32, max_seq_len=1024,
                 store_slots=65536, d_k=128, d_v=256, pq_m=4, pq_k=256, topk=8, dedup_threshold=0.97,
                 softcap=30.0, rope_theta=10000.0, local_memory=True, grad_checkpoint=False,
                 growth_history=None, **kw):
        if block_types is None:
            block_types = ["attn" if i in (3, 7, 11) else "lin" for i in range(12)]
        assert d_model == n_heads * head_dim, "d_model must equal n_heads*head_dim"
        assert n_heads % n_kv_heads == 0 and d_v % pq_m == 0 and pq_k <= 256
        self.vocab_size, self.d_model, self.n_heads, self.n_kv_heads, self.head_dim = vocab_size, d_model, n_heads, n_kv_heads, head_dim
        self.ffn_hidden, self.block_types = ffn_hidden, list(block_types)
        self.read_after, self.write_from = [int(x) for x in read_after], int(write_from)
        self.rec_chunk, self.write_chunk, self.max_seq_len = rec_chunk, write_chunk, max_seq_len
        self.store_slots, self.d_k, self.d_v, self.pq_m, self.pq_k, self.topk = store_slots, d_k, d_v, pq_m, pq_k, topk
        self.dedup_threshold, self.softcap, self.rope_theta = dedup_threshold, softcap, rope_theta
        self.local_memory, self.grad_checkpoint = local_memory, grad_checkpoint
        self.growth_history = list(growth_history or [])
        self.n_layers = len(self.block_types)
        kw["tie_word_embeddings"] = False  # tying is done manually (embedding matrix is reused as the LM head)
        kw.setdefault("pad_token_id", 0)
        kw.setdefault("eos_token_id", 1)
        super().__init__(**kw)


# ----------------------------------------------------------------------------------------------
# Backbone
# ----------------------------------------------------------------------------------------------
class RMSNorm(nn.Module):
    def __init__(self, d, eps=1e-8):
        super().__init__()
        self.weight, self.eps = nn.Parameter(torch.ones(d)), eps

    def forward(self, x):
        xf = x.float()
        return (xf * torch.rsqrt(xf.pow(2).mean(-1, keepdim=True) + self.eps)).to(x.dtype) * self.weight


def rope(x, theta):  # x: (B,H,L,Dh)
    half = x.shape[-1] // 2
    freqs = 1.0 / (theta ** (torch.arange(half, device=x.device, dtype=torch.float32) / half))
    ang = torch.arange(x.shape[2], device=x.device, dtype=torch.float32)[:, None] * freqs[None]
    cos, sin = ang.cos(), ang.sin()
    x1, x2 = x[..., :half].float(), x[..., half:].float()
    return torch.cat([x1 * cos - x2 * sin, x1 * sin + x2 * cos], -1).to(x.dtype)


class SwiGLU(nn.Module):
    def __init__(self, d, h):
        super().__init__()
        self.w1, self.w3, self.w2 = nn.Linear(d, h, bias=False), nn.Linear(d, h, bias=False), nn.Linear(h, d, bias=False)

    def forward(self, x):
        return self.w2(F.silu(self.w1(x)) * self.w3(x))


def chunked_gated_linear_attention(q, k, v, g, chunk):
    """q,k,v: (B,L,H,dh) float32 (q,k L2-normalised); g: (B,L,H) log-decay <= 0.  Returns (B,L,H,dh).
    Within a chunk: decay-masked matmul.  Between chunks: carried state S (B,H,dh,dh).  All in fp32, log-space decay."""
    B, L, H, dh = q.shape
    pad = (-L) % chunk
    if pad:
        q, k, v = [F.pad(t, (0, 0, 0, 0, 0, pad)) for t in (q, k, v)]
        g = F.pad(g, (0, 0, 0, pad))
    Lp = L + pad
    nc = Lp // chunk
    q, k, v = [t.reshape(B, nc, chunk, H, dh).permute(0, 3, 1, 2, 4) for t in (q, k, v)]  # B,H,nc,C,dh
    g = g.reshape(B, nc, chunk, H).permute(0, 3, 1, 2)  # B,H,nc,C
    b = g.cumsum(-1)
    mask = torch.tril(torch.ones(chunk, chunk, dtype=torch.bool, device=q.device))
    diff = (b.unsqueeze(-1) - b.unsqueeze(-2)).masked_fill(~mask, float("-inf"))
    intra = torch.matmul(torch.matmul(q, k.transpose(-1, -2)) * torch.exp(diff), v)
    qd = q * torch.exp(b).unsqueeze(-1)
    kd = k * torch.exp(b[..., -1:] - b).unsqueeze(-1)
    total = torch.exp(b[..., -1])
    S = q.new_zeros(B, H, dh, dh)
    outs = []
    for c in range(nc):
        outs.append(torch.matmul(qd[:, :, c], S))
        S = S * total[:, :, c, None, None] + torch.matmul(kd[:, :, c].transpose(-1, -2), v[:, :, c])
    o = intra + torch.stack(outs, 2)  # B,H,nc,C,dh
    return o.permute(0, 2, 3, 1, 4).reshape(B, Lp, H, dh)[:, :L]


def reference_recurrence(q, k, v, g):
    B, L, H, dh = q.shape
    S = q.new_zeros(B, H, dh, dh)
    out = []
    for t in range(L):
        S = torch.exp(g[:, t])[..., None, None] * S + k[:, t].unsqueeze(-1) * v[:, t].unsqueeze(-2)
        out.append(torch.einsum("bhd,bhde->bhe", q[:, t], S))
    return torch.stack(out, 1)


class LinearRecurrentMix(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        D, H, dh = cfg.d_model, cfg.n_heads, cfg.head_dim
        self.H, self.dh, self.chunk = H, dh, cfg.rec_chunk
        self.q, self.k, self.v = (nn.Linear(D, H * dh, bias=False) for _ in range(3))
        self.g = nn.Linear(D, H)
        self.o_gate = nn.Linear(D, H * dh, bias=False)
        self.onorm = RMSNorm(dh)
        self.out = nn.Linear(H * dh, D, bias=False)

    def forward(self, x):
        B, L, _ = x.shape
        H, dh = self.H, self.dh
        og = torch.sigmoid(self.o_gate(x)).view(B, L, H, dh)
        with torch.autocast(device_type=x.device.type, enabled=False):
            q = F.normalize(self.q(x).float().view(B, L, H, dh), dim=-1)
            k = F.normalize(self.k(x).float().view(B, L, H, dh), dim=-1)
            v = self.v(x).float().view(B, L, H, dh)
            g = -F.softplus(self.g(x).float())
            o = chunked_gated_linear_attention(q, k, v, g, self.chunk)
        o = self.onorm(o.to(x.dtype)) * og
        return self.out(o.reshape(B, L, H * dh))


class CausalAttention(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        D, H, Hkv, dh = cfg.d_model, cfg.n_heads, cfg.n_kv_heads, cfg.head_dim
        self.H, self.Hkv, self.dh, self.theta = H, Hkv, dh, cfg.rope_theta
        self.q = nn.Linear(D, H * dh, bias=False)
        self.k = nn.Linear(D, Hkv * dh, bias=False)
        self.v = nn.Linear(D, Hkv * dh, bias=False)
        self.out = nn.Linear(H * dh, D, bias=False)
        self.qn, self.kn = RMSNorm(dh), RMSNorm(dh)

    def forward(self, x):
        B, L, _ = x.shape
        q = self.qn(self.q(x).view(B, L, self.H, self.dh)).transpose(1, 2)
        k = self.kn(self.k(x).view(B, L, self.Hkv, self.dh)).transpose(1, 2)
        v = self.v(x).view(B, L, self.Hkv, self.dh).transpose(1, 2)
        q, k = rope(q, self.theta), rope(k, self.theta)
        g = self.H // self.Hkv
        k, v = k.repeat_interleave(g, 1), v.repeat_interleave(g, 1)
        o = F.scaled_dot_product_attention(q, k, v, is_causal=True)
        return self.out(o.transpose(1, 2).reshape(B, L, self.H * self.dh))


class Block(nn.Module):
    def __init__(self, cfg, kind):
        super().__init__()
        self.kind = kind
        self.n1, self.n2 = RMSNorm(cfg.d_model), RMSNorm(cfg.d_model)
        self.mix = CausalAttention(cfg) if kind == "attn" else LinearRecurrentMix(cfg)
        self.ffn = SwiGLU(cfg.d_model, cfg.ffn_hidden)

    def forward(self, x):
        x = x + self.mix(self.n1(x))
        return x + self.ffn(self.n2(x))


# ----------------------------------------------------------------------------------------------
# Barameter memory (numpy-backed, scalable)
# ----------------------------------------------------------------------------------------------
class GrowArray:
    """Append-only numpy array with doubling capacity; can be opened as a read-only memmap."""

    def __init__(self, tail, dtype, cap=1024):
        self.tail, self.dtype = tuple(tail), np.dtype(dtype)
        self.buf, self.n, self.ro = np.zeros((cap,) + self.tail, self.dtype), 0, False

    @property
    def data(self):
        return self.buf[: self.n]

    def append(self, arr):
        if self.ro:
            raise RuntimeError("memory opened read-only (--mmap); reload without --mmap to modify it")
        arr = np.asarray(arr, self.dtype)
        m, need = len(arr), self.n + len(arr)
        if need > len(self.buf):
            nb = np.zeros((max(need, 2 * len(self.buf)),) + self.tail, self.dtype)
            nb[: self.n] = self.buf[: self.n]
            self.buf = nb
        st = self.n
        self.buf[st:st + m] = arr
        self.n += m
        return st

    def keep(self, mask):
        d = np.array(self.buf[: self.n][mask])
        self.buf = np.concatenate([d, np.zeros((16,) + self.tail, self.dtype)])
        self.n, self.ro = len(d), False

    def save(self, path):
        np.save(path, np.ascontiguousarray(self.data))

    def load(self, path, mmap=False):
        a = np.load(path, mmap_mode="r" if mmap else None)
        self.buf, self.n, self.ro = (a, len(a), True) if mmap else (np.concatenate([a, np.zeros((16,) + self.tail, self.dtype)]), len(a), False)


def chunked_topk(keys, q, k, step=16384):
    """Brute-force top-k of q (Q,dk) against keys (N,dk) (torch or numpy/memmap), streaming over N."""
    N = len(keys)
    if N == 0:
        return q.new_zeros(len(q), 0), torch.zeros(len(q), 0, dtype=torch.long, device=q.device)
    best_s = best_i = None
    for a in range(0, N, step):
        kc = keys[a:a + step]
        if isinstance(kc, np.ndarray):
            kc = torch.from_numpy(np.ascontiguousarray(kc))
        sc = q @ kc.to(q.device).to(q.dtype).T
        s, i = sc.topk(min(k, sc.shape[1]), dim=1)
        i = i + a
        if best_s is None:
            best_s, best_i = s, i
        else:
            cs, ci = torch.cat([best_s, s], 1), torch.cat([best_i, i], 1)
            best_s, j = cs.topk(min(k, cs.shape[1]), 1)
            best_i = ci.gather(1, j)
    return best_s.float(), best_i


def hash_rows(tokens):
    return np.array([int.from_bytes(hashlib.blake2b(np.ascontiguousarray(r).tobytes(), digest_size=8).digest(), "little", signed=True)
                     for r in tokens], np.int64)


class _KeyedArrays:
    """Shared helpers: GPU key cache and sorted-hash lookup."""
    _cache = None
    _hs = None

    def _qdtype(self, q):
        return q.half() if q.is_cuda else q.float()

    def keys_for(self, device):
        if device.type == "cuda" and self.keys.n * self.keys.tail[0] * 2 < 1.5e9:
            if self._cache is None or len(self._cache) != self.keys.n:
                self._cache = torch.from_numpy(np.ascontiguousarray(self.keys.data)).to(device)
            return self._cache
        return self.keys.data

    def search(self, q, k):
        return chunked_topk(self.keys_for(q.device), self._qdtype(q), k)

    def lookup(self, h):
        if self._hs is None:
            order = np.argsort(self.hash.data, kind="stable")
            self._hs, self._ho = self.hash.data[order], order
        if len(self._hs) == 0:
            return np.zeros(len(h), np.int64), np.zeros(len(h), bool)
        pos = np.minimum(np.searchsorted(self._hs, h), len(self._hs) - 1)
        return self._ho[pos], self._hs[pos] == h

    def invalidate(self, hashes_changed=False):
        self._cache = None
        if hashes_changed:
            self._hs = None


class ChunkIndex(_KeyedArrays):
    """Cold tier: every chunk ever seen (key + tokens). Never evicts; used by the deep search path."""

    def __init__(self, cfg):
        self.keys = GrowArray((cfg.d_k,), np.float16)
        self.tok = GrowArray((cfg.write_chunk,), np.int32)
        self.ds = GrowArray((), np.int32)
        self.hash = GrowArray((), np.int64)

    @property
    def n(self):
        return self.keys.n

    def add(self, keys, tokens, ds):
        tokens = np.ascontiguousarray(tokens, np.int32)
        h, keys = hash_rows(tokens), np.asarray(keys, np.float16)
        slot, hit = self.lookup(h)
        rows = np.zeros(len(h), np.int64)
        rows[hit] = slot[hit]
        self.keys.buf[rows[hit]] = keys[hit]  # refresh with the newest key
        new = ~hit
        if new.any():
            st = self.keys.append(keys[new])
            self.tok.append(tokens[new]); self.ds.append(np.full(int(new.sum()), ds, np.int32)); self.hash.append(h[new])
            rows[new] = np.arange(st, st + int(new.sum()))
            self.invalidate(True)
        else:
            self.invalidate()
        return rows, h

    def delete_ds(self, ds):
        mask = self.ds.data != ds
        remap = np.where(mask, np.cumsum(mask) - 1, -1).astype(np.int64)
        for a in (self.keys, self.tok, self.ds, self.hash):
            a.keep(mask)
        self.invalidate(True)
        return int((~mask).sum()), remap

    def window(self, w, per):
        return self.tok.data[w * per:(w + 1) * per].reshape(-1)

    def arrays(self):
        return {"keys": self.keys, "tok": self.tok, "ds": self.ds, "hash": self.hash}


class BarameterStore(_KeyedArrays):
    """Hot tier: fixed-capacity slots. key (L2-normalised fp16) + 4x8-bit PQ codes + metadata."""

    def __init__(self, cfg):
        self.cfg, self.cap, self.clock = cfg, cfg.store_slots, 0
        self.keys = GrowArray((cfg.d_k,), np.float16, 4096)
        self.codes = GrowArray((cfg.pq_m,), np.uint8, 4096)
        self.ds = GrowArray((), np.int32, 4096)
        self.usage = GrowArray((), np.int32, 4096)
        self.step = GrowArray((), np.int64, 4096)
        self.chunk = GrowArray((), np.int64, 4096)  # row in ChunkIndex (source text pointer)
        self.hash = GrowArray((), np.int64, 4096)
        self._codes_cache = None

    @property
    def n(self):
        return self.keys.n

    def arrays(self):
        return {"keys": self.keys, "codes": self.codes, "ds": self.ds, "usage": self.usage, "step": self.step,
                "chunk": self.chunk, "hash": self.hash}

    def invalidate(self, hashes_changed=False):
        super().invalidate(hashes_changed)
        self._codes_cache = None

    def _keep(self, mask):
        for a in self.arrays().values():
            a.keep(mask)
        self.invalidate(True)

    def gather_keys(self, idx, device):
        kf = self.keys_for(device)
        if isinstance(kf, torch.Tensor):
            return kf[idx].float()
        flat = np.asarray(kf[idx.cpu().numpy().reshape(-1)])
        return torch.from_numpy(flat).to(device).float().view(*idx.shape, -1)

    def dequant(self, idx, codebook):
        if self._codes_cache is None or len(self._codes_cache) != self.n or self._codes_cache.device != idx.device:
            self._codes_cache = torch.from_numpy(np.ascontiguousarray(self.codes.data)).to(idx.device)
        c = self._codes_cache[idx].long()  # (...,m)
        ar = torch.arange(c.shape[-1], device=idx.device)
        return codebook[ar, c].reshape(*idx.shape, -1)

    def write(self, keys, codes, ds, rows, hashes):
        keys = F.normalize(torch.as_tensor(keys).float(), dim=-1)
        k16 = keys.numpy().astype(np.float16)
        codes = np.asarray(codes, np.uint8)
        self.clock += 1
        slot, hit = self.lookup(hashes)
        for i in np.nonzero(hit)[0]:  # same chunk again: refresh in place
            j = slot[i]
            self.keys.buf[j], self.codes.buf[j] = k16[i], codes[i]
            self.usage.buf[j] += 1
            self.step.buf[j] = self.clock
        todo = np.nonzero(~hit)[0]
        if len(todo) and self.n > 0:  # near-duplicate merge
            sc, ix = self.search(keys[todo], 1)
            sims, ix = sc[:, 0].cpu().numpy(), ix[:, 0].cpu().numpy()
            merge = sims >= self.cfg.dedup_threshold
            for t, j in zip(todo[merge], ix[merge]):
                nk = 0.9 * self.keys.buf[j].astype(np.float32) + 0.1 * keys[t].numpy()
                self.keys.buf[j] = (nk / (np.linalg.norm(nk) + 1e-8)).astype(np.float16)
                self.usage.buf[j] += 1
                self.step.buf[j] = self.clock
            todo = todo[~merge]
        if len(todo):
            todo = todo[-self.cap:]
            need = len(todo) - (self.cap - self.n)
            if need > 0:  # evict least-used, then oldest
                order = np.lexsort((self.step.data, self.usage.data))[:need]
                mask = np.ones(self.n, bool)
                mask[order] = False
                self._keep(mask)
            m = len(todo)
            self.keys.append(k16[todo]); self.codes.append(codes[todo]); self.ds.append(np.full(m, ds, np.int32))
            self.usage.append(np.ones(m, np.int32)); self.step.append(np.full(m, self.clock, np.int64))
            self.chunk.append(rows[todo]); self.hash.append(hashes[todo])
        self.invalidate(True)

    def delete_ds(self, ds):
        mask = self.ds.data != ds
        n = int((~mask).sum())
        if n:
            self._keep(mask)
        return n

    def remap_chunks(self, remap):
        c = self.chunk.data
        ok = c >= 0
        self.chunk.buf[: self.n][ok] = remap[c[ok]]
        gone = self.chunk.data < 0
        if gone.any():
            self._keep(~gone)

    def clear(self):
        self._keep(np.zeros(self.n, bool))


class Memory:
    """Hot store + cold index + dataset tag registry. Lives on the model as model.mem (not an nn.Module)."""

    def __init__(self, cfg):
        self.cfg, self.datasets = cfg, []
        self.store, self.index = BarameterStore(cfg), ChunkIndex(cfg)

    def ds_id(self, tag):
        if tag not in self.datasets:
            self.datasets.append(tag)
        return self.datasets.index(tag)

    def write(self, keys, codes, tokens, tag):
        ds = self.ds_id(tag)
        rows, h = self.index.add(torch.as_tensor(keys).float().numpy(), tokens, ds)
        self.store.write(keys, codes, ds, rows, h)

    def delete(self, tag):
        if tag not in self.datasets:
            return 0, 0
        ds = self.datasets.index(tag)
        ns = self.store.delete_ds(ds)
        ni, remap = self.index.delete_ds(ds)
        self.store.remap_chunks(remap)
        return ns, ni

    def counts(self):
        out = {}
        for i, t in enumerate(self.datasets):
            out[t] = (int((self.store.ds.data == i).sum()), int((self.index.ds.data == i).sum()))
        return out

    def save(self, d):
        os.makedirs(d, exist_ok=True)
        for part, obj in (("store", self.store), ("index", self.index)):
            for name, arr in obj.arrays().items():
                arr.save(os.path.join(d, f"{part}_{name}.npy"))
        json.dump({"datasets": self.datasets, "clock": self.store.clock}, open(os.path.join(d, "memory.json"), "w"))

    def load(self, d, mmap=False):
        if not os.path.exists(os.path.join(d, "memory.json")):
            return
        meta = json.load(open(os.path.join(d, "memory.json")))
        self.datasets, self.store.clock = meta["datasets"], meta["clock"]
        for part, obj in (("store", self.store), ("index", self.index)):
            for name, arr in obj.arrays().items():
                arr.load(os.path.join(d, f"{part}_{name}.npy"), mmap)
            obj.invalidate(True)


# ----------------------------------------------------------------------------------------------
# Barameter read / write modules
# ----------------------------------------------------------------------------------------------
class BarameterWriter(nn.Module):
    """Text chunk -> (key, PQ-coded value). Attention-pool hidden states, project, product-quantise (straight-through)."""

    def __init__(self, cfg):
        super().__init__()
        D, m = cfg.d_model, cfg.pq_m
        self.m, self.K, self.ds_, self.dk, self.dv = m, cfg.pq_k, cfg.d_v // m, cfg.d_k, cfg.d_v
        self.norm = RMSNorm(D)
        self.pool_q = nn.Parameter(torch.randn(D) * 0.02)
        self.wk, self.wv = nn.Linear(D, cfg.d_k, bias=False), nn.Linear(D, cfg.d_v, bias=False)
        self.codebook = nn.Parameter(torch.randn(m, cfg.pq_k, self.ds_) * 0.5)
        self.register_buffer("usage", torch.zeros(m, cfg.pq_k))
        self.last_vs = None

    def forward(self, h, C):
        B, L, D = h.shape
        nc = L // C
        if nc == 0:
            return None
        hh = self.norm(h[:, : nc * C]).float().view(B, nc, C, D)
        a = torch.softmax((hh * self.pool_q.float()).sum(-1) * 0.125, -1)  # fixed temperature: independent of d_model (growth-safe)
        pooled = (a.unsqueeze(-1) * hh).sum(2)
        with torch.autocast(device_type=h.device.type, enabled=False):
            key = F.normalize(self.wk(pooled), dim=-1)
            vs = self.wv(pooled).view(B, nc, self.m, self.ds_)
            cb = self.codebook.float()
            dist = (vs ** 2).sum(-1, keepdim=True) - 2 * torch.einsum("bcmd,mkd->bcmk", vs, cb) + (cb ** 2).sum(-1)
            codes = dist.argmin(-1)  # B,nc,m
            qv = cb[torch.arange(self.m, device=h.device), codes]  # B,nc,m,ds
            commit = 0.25 * F.mse_loss(vs, qv.detach()) + F.mse_loss(qv, vs.detach())
            vq = (vs + (qv - vs).detach()).reshape(B, nc, self.dv)
        div = key.reshape(-1, self.dk).mean(0).pow(2).sum()
        if self.training:
            with torch.no_grad():
                self.usage += F.one_hot(codes, self.K).sum((0, 1)).float()
                self.last_vs = vs.detach().reshape(-1, self.m, self.ds_)[:4096]
        return {"keys": key, "vals": vq, "codes": codes, "aux": {"commit": commit, "div": div}}

    @torch.no_grad()
    def reinit_dead(self):
        if self.last_vs is None:
            return 0
        n = 0
        for mi in range(self.m):
            dead = (self.usage[mi] == 0).nonzero().flatten()
            if len(dead):
                pick = self.last_vs[torch.randint(len(self.last_vs), (len(dead),), device=self.last_vs.device), mi]
                self.codebook.data[mi, dead] = pick + 0.01 * torch.randn_like(pick)
                n += len(dead)
        self.usage.zero_()
        return n


class BarameterRead(nn.Module):
    """x += sigmoid(gate) * Wo( softmax(top-k persistent slots + causal in-sequence chunks) . values ). Starts ~off."""

    def __init__(self, cfg):
        super().__init__()
        D = cfg.d_model
        self.cfg = cfg
        self.norm = RMSNorm(D)
        self.wq = nn.Linear(D, cfg.d_k, bias=False)
        self.gate = nn.Linear(D, 1)
        self.wo = nn.Linear(cfg.d_v, D, bias=False)
        self.log_scale = nn.Parameter(torch.tensor(math.log(10.0)))
        nn.init.constant_(self.gate.bias, -4.0)

    def forward(self, x, mem, codebook, local, use_store):
        cfg = self.cfg
        B, L, _ = x.shape
        h = self.norm(x)
        q = F.normalize(self.wq(h).float(), dim=-1)
        scale = self.log_scale.exp().clamp(max=64.0)
        parts, sp, vp = [], None, None
        if use_store and mem.store.n > 0:
            qf = q.reshape(B * L, -1)
            with torch.no_grad():
                _, idx = mem.store.search(qf.detach(), cfg.topk)
            kk = mem.store.gather_keys(idx, x.device)
            sp = (scale * (qf.unsqueeze(1) * kk).sum(-1)).view(B, L, -1)
            vp = mem.store.dequant(idx, codebook).float().view(B, L, idx.shape[1], -1)
            parts.append(sp)
        if local is not None:
            lk, C = local["keys"], cfg.write_chunk
            nc = lk.shape[1]
            sl = scale * torch.einsum("bld,bcd->blc", q, lk)
            valid = (torch.arange(L, device=x.device) // C)[:, None] > torch.arange(nc, device=x.device)[None]
            parts.append(sl.masked_fill(~valid[None], -1e4))
        if not parts:
            return x, q
        allsc = torch.cat(parts, -1)
        w = torch.softmax(allsc, -1) * (allsc > -1e3)
        r, off = 0, 0
        if sp is not None:
            kp = sp.shape[-1]
            r = r + torch.einsum("blk,blkd->bld", w[..., :kp], vp)
            off = kp
        if local is not None:
            r = r + torch.einsum("blc,bcd->bld", w[..., off:], local["vals"].float())
        out = torch.sigmoid(self.gate(h).float()) * self.wo(r.to(h.dtype)).float()
        return x + out.to(x.dtype), q


# ----------------------------------------------------------------------------------------------
# Model
# ----------------------------------------------------------------------------------------------
class BarameterForCausalLM(PreTrainedModel):
    config_class = BarameterConfig
    base_model_prefix = "barameter"
    _tied_weights_keys = []

    def __init__(self, config):
        super().__init__(config)
        self.all_tied_weights_keys = {}
        cfg = config
        self.embed = nn.Embedding(cfg.vocab_size, cfg.d_model)
        self.blocks = nn.ModuleList([Block(cfg, t) for t in cfg.block_types])
        self.reads = nn.ModuleDict({str(n): BarameterRead(cfg) for n in cfg.read_after})
        self.writer = BarameterWriter(cfg)
        self.norm = RMSNorm(cfg.d_model)
        self.mem = Memory(cfg)
        self.last_writes, self.last_aux = None, {}
        self._init()

    def _init(self):
        std, out_std = 0.02, 0.02 / math.sqrt(2 * self.config.n_layers)
        for name, m in self.named_modules():
            if isinstance(m, nn.Linear):
                tail = name.split(".")[-1]
                nn.init.normal_(m.weight, 0, out_std if tail in ("out", "w2", "wo") else std)
                if m.bias is not None:
                    nn.init.zeros_(m.bias)
            elif isinstance(m, nn.Embedding):
                nn.init.normal_(m.weight, 0, std)
        for b in self.blocks:
            if b.kind == "lin":
                with torch.no_grad():
                    b.mix.g.bias.copy_(torch.linspace(-4.5, -1.0, b.mix.H))
        for n, r in self.reads.items():
            nn.init.constant_(r.gate.bias, -4.0)

    def get_input_embeddings(self):
        return self.embed

    def num_params(self):
        return sum(p.numel() for p in self.parameters())

    # ---- core ----
    def _run(self, ids, use_store=True, C=None, stop_at_write=False):
        cfg = self.config
        C = C or cfg.write_chunk
        x, writes, aux = self.embed(ids), None, {}
        for i, blk in enumerate(self.blocks):
            if cfg.grad_checkpoint and self.training and torch.is_grad_enabled():
                x = checkpoint(blk, x, use_reentrant=False)
            else:
                x = blk(x)
            n = i + 1
            if n == cfg.write_from:
                writes = self.writer(x, C)
                if writes is not None:
                    aux.update(writes["aux"])
                if stop_at_write:
                    return x, writes, aux
            if str(n) in self.reads:
                local = writes if (writes is not None and cfg.local_memory and n > cfg.write_from) else None
                rd = self.reads[str(n)]
                x, q = rd(x, self.mem, self.writer.codebook, local, use_store)
                if local is not None and self.training and local["keys"].shape[1] > 1:
                    nc, B = local["keys"].shape[1], ids.shape[0]
                    qm = F.normalize(q[:, : nc * C].reshape(B, nc, C, -1).mean(2), dim=-1)[:, 1:].reshape(B * (nc - 1), -1)
                    kt = local["keys"][:, :-1].reshape(B * (nc - 1), -1)
                    lg = rd.log_scale.exp().clamp(max=64.0) * qm @ kt.T
                    aux["contrast"] = F.cross_entropy(lg, torch.arange(len(lg), device=lg.device))
        return x, writes, aux

    def forward(self, input_ids, labels=None, use_store=True, reduction="mean", **kw):
        cfg = self.config
        x, writes, aux = self._run(input_ids, use_store)
        h = self.norm(x)
        logits = cfg.softcap * torch.tanh(F.linear(h, self.embed.weight.to(h.dtype)).float() / cfg.softcap)
        loss = None
        if labels is not None:
            lg, tg = logits[:, :-1], labels[:, 1:]
            ce = F.cross_entropy(lg.reshape(-1, lg.shape[-1]), tg.reshape(-1), ignore_index=-100, reduction="none").view(tg.shape)
            valid = (tg != -100).float()
            if reduction == "none":
                loss = (ce * valid).sum(1) / valid.sum(1).clamp(min=1)
            else:
                ce_m = (ce * valid).sum() / valid.sum().clamp(min=1)
                z = torch.logsumexp(lg, -1).pow(2).mean()
                loss = ce_m + 1e-4 * z + aux.get("commit", 0) + 0.1 * aux.get("contrast", 0) + 0.01 * aux.get("div", 0)
                self.last_aux = {"ce": ce_m.detach(), **{k: v.detach() for k, v in aux.items()}}
        self.last_writes = writes
        return CausalLMOutput(loss=loss, logits=logits)

    @torch.no_grad()
    def encode_chunks(self, ids, chunk=None):
        """Forward-only barameter writer (ingestion path). Returns dict(keys, codes) or None."""
        was = self.training
        self.eval()
        _, writes, _ = self._run(ids, True, chunk, stop_at_write=True)
        self.train(was)
        return writes

    @torch.no_grad()
    def chat_generate(self, ids, max_new=128, temperature=0.7, top_k=40, eos=1):
        self.eval()
        out = ids.clone()
        for _ in range(max_new):
            lg = self(out[:, -self.config.max_seq_len:]).logits[:, -1].float()
            if temperature <= 0:
                nxt = lg.argmax(-1, keepdim=True)
            else:
                lg = lg / temperature
                v, _ = lg.topk(min(top_k, lg.shape[-1]))
                lg = lg.masked_fill(lg < v[:, -1:], float("-inf"))
                nxt = torch.multinomial(torch.softmax(lg, -1), 1)
            out = torch.cat([out, nxt], 1)
            if int(nxt) == eos:
                break
        return out

    def load_memory(self, path_or_repo, mmap=False):
        d = resolve_path(path_or_repo)
        self.mem.load(os.path.join(d, "memory"), mmap)


try:
    AutoConfig.register("barameter", BarameterConfig)
except Exception:
    pass
try:
    AutoModelForCausalLM.register(BarameterConfig, BarameterForCausalLM)
except Exception:
    pass


# ----------------------------------------------------------------------------------------------
# Function-preserving growth (depth and width) - no pretraining restart
# ----------------------------------------------------------------------------------------------
ZERO_SUFFIX = ("mix.out.weight", "ffn.w2.weight", "wo.weight")


def grow_model(model, add_layers=0, new_d_model=None):
    old = model.config
    cfg = copy.deepcopy(old)
    D0, dh, g = old.d_model, old.head_dim, old.n_heads // old.n_kv_heads
    if new_d_model and new_d_model > D0:
        H1 = math.ceil(new_d_model / dh / g) * g  # keeps GQA group size so old heads keep their KV mapping
        cfg.d_model, cfg.n_heads, cfg.n_kv_heads = H1 * dh, H1, H1 // g
        cfg.ffn_hidden = int(round(old.ffn_hidden * cfg.d_model / D0 / 64)) * 64
    D1 = cfg.d_model
    n_old = old.n_layers
    ins = sorted({min(n_old - 1, max(1, round((j + 1) * n_old / (add_layers + 1)))) for j in range(add_layers)}) if add_layers else []
    types, new2old, oi = [], {}, 0
    for pos in range(n_old):
        while ins and ins[0] == pos:
            types.append("lin"); ins.pop(0)
        new2old[len(types)] = pos
        types.append(old.block_types[pos])
    while ins:
        types.append("lin"); ins.pop(0)
    inserted_before = lambda c: sum(1 for ni in range(len(types)) if ni not in new2old and ni < _new_index_of_old(new2old, c))
    old2new = {o: n for n, o in new2old.items()}
    shift = lambda c: (old2new[c - 1] + 1) if c >= 1 else 0  # count after old block c-1
    cfg.block_types, cfg.n_layers = types, len(types)
    cfg.read_after = [shift(c) for c in old.read_after]
    cfg.write_from = shift(old.write_from)
    cfg.growth_history = list(old.growth_history) + [{"add_layers": add_layers, "d_model": [D0, D1], "layers": [n_old, len(types)]}]
    new = BarameterForCausalLM(cfg).to(next(model.parameters()).device)
    read_new2old = {shift(c): c for c in old.read_after}
    osd = model.state_dict()

    def old_name(nm):
        p = nm.split(".")
        if p[0] == "blocks":
            if int(p[1]) not in new2old:
                return None
            p[1] = str(new2old[int(p[1])])
        elif p[0] == "reads":
            p[1] = str(read_new2old[int(p[1])])
        return ".".join(p)

    with torch.no_grad():
        for name, p in new.state_dict().items():
            on = old_name(name)
            if name.endswith(ZERO_SUFFIX):
                p.zero_()
            if on is None:
                continue
            src = osd[on].to(p.dtype)
            if name == "embed.weight":
                p.zero_()
            tail = ".".join(name.split(".")[-2:])
            if tail in ("n1.weight", "n2.weight", "norm.weight") and D1 != D0:
                p.fill_(1.0)
                src = src * math.sqrt(D0 / D1)  # zero-padded dims change RMS; compensate so outputs are identical
            assert all(a >= b for a, b in zip(p.shape, src.shape)), name
            p[tuple(slice(0, s) for s in src.shape)] = src
    new.mem = model.mem
    new.mem.cfg = cfg
    return new


def _new_index_of_old(new2old, c):
    return 0


# ----------------------------------------------------------------------------------------------
# Tokenizer, data
# ----------------------------------------------------------------------------------------------
class Tok:
    pad, eos, user, asst = 0, 1, 2, 3

    def __init__(self, enc):
        self.enc = enc

    @property
    def vocab_size(self):
        return self.enc.get_vocab_size()

    def encode(self, t):
        return self.enc.encode(t).ids

    def decode(self, ids):
        return self.enc.decode([int(i) for i in ids], skip_special_tokens=False)

    def save(self, d):
        os.makedirs(d, exist_ok=True)
        self.enc.save(os.path.join(d, "tokenizer.json"))
        try:
            from transformers import PreTrainedTokenizerFast
            PreTrainedTokenizerFast(tokenizer_object=self.enc, pad_token="<pad>", eos_token="<eos>").save_pretrained(d)
        except Exception:
            pass

    @staticmethod
    def load(d):
        from tokenizers import Tokenizer
        return Tok(Tokenizer.from_file(os.path.join(d, "tokenizer.json")))

    @staticmethod
    def train(texts, vocab_size):
        from tokenizers import Tokenizer, models, pre_tokenizers, decoders, trainers
        enc = Tokenizer(models.BPE())
        enc.pre_tokenizer = pre_tokenizers.ByteLevel(add_prefix_space=False)
        enc.decoder = decoders.ByteLevel()
        tr = trainers.BpeTrainer(vocab_size=vocab_size, special_tokens=SPECIALS, show_progress=False,
                                 initial_alphabet=pre_tokenizers.ByteLevel.alphabet())
        enc.train_from_iterator(texts, tr)
        return Tok(enc)


def format_row(r):
    if "instruction" in r:
        u = r["instruction"] + (("\n" + r["input"]) if r.get("input") else "")
        a = r.get("output") or r.get("response")
    elif "question" in r:
        u, a = r["question"], r.get("response") or r.get("answer")
    elif "prompt" in r:
        u, a = r["prompt"], r.get("response") or r.get("completion") or r.get("chosen")
    elif "text" in r:
        return r["text"]
    else:
        return None
    if not u or not a:
        return None
    return f"<|user|>\n{u}\n<|assistant|>\n{a}"


def load_rows(name, split="train", streaming=False, max_examples=0, holdout=500):
    """Returns (make_rows(epoch) -> iterator of strings, heldout list of strings)."""
    from datasets import load_dataset
    ds = load_dataset(name, split=split, streaming=streaming)
    if streaming:
        held = [t for t in (format_row(r) for r in ds.take(holdout)) if t]
        tr = ds.skip(holdout)
        if max_examples:
            tr = tr.take(max_examples)
    else:
        n = len(ds)
        held = [t for t in (format_row(ds[i]) for i in range(min(holdout, n))) if t]
        tr = ds.select(range(min(holdout, n), n if not max_examples else min(n, holdout + max_examples)))

    def make_rows(epoch):
        src = tr.shuffle(seed=epoch, buffer_size=10000) if streaming else tr.shuffle(seed=epoch)
        for r in src:
            t = format_row(r)
            if t:
                yield t

    return make_rows, held


def _pack(tok, texts, L):
    buf, batch = [], []

    def flush():
        for e in tok.enc.encode_batch(batch):
            buf.extend(e.ids)
            buf.append(tok.eos)
        batch.clear()

    for t in texts:
        batch.append(t)
        if len(batch) >= 256:
            flush()
            while len(buf) >= L:
                yield np.asarray(buf[:L], np.int64)
                del buf[:L]
    if batch:
        flush()
    while len(buf) >= L:
        yield np.asarray(buf[:L], np.int64)
        del buf[:L]


def window_stream(tok, make_rows, L, skip=0):
    ep, n = 0, 0
    while True:
        got = False
        for w in _pack(tok, make_rows(ep), L):
            got = True
            n += 1
            if n > skip:
                yield w
        if not got:
            return
        ep += 1


# ----------------------------------------------------------------------------------------------
# Optimiser (Muon for 2-D hidden weights, AdamW for the rest)
# ----------------------------------------------------------------------------------------------
def newton_schulz(G, steps=5):
    a, b, c = 3.4445, -4.7750, 2.0315
    X = G.float()
    X = X / (X.norm() + 1e-7)
    tr = X.shape[0] > X.shape[1]
    if tr:
        X = X.T
    for _ in range(steps):
        A = X @ X.T
        X = a * X + (b * A + c * A @ A) @ X
    return X.T if tr else X


class Muon(torch.optim.Optimizer):
    def __init__(self, params, lr=0.02, momentum=0.95, wd=0.01):
        super().__init__(params, dict(lr=lr, momentum=momentum, wd=wd))

    @torch.no_grad()
    def step(self, closure=None):
        for g in self.param_groups:
            for p in g["params"]:
                if p.grad is None:
                    continue
                st = self.state[p]
                if "buf" not in st:
                    st["buf"] = torch.zeros_like(p)
                st["buf"].mul_(g["momentum"]).add_(p.grad)
                u = newton_schulz(p.grad.add(st["buf"], alpha=g["momentum"]))
                u = u * max(1.0, p.shape[0] / p.shape[1]) ** 0.5
                p.mul_(1 - g["lr"] * g["wd"]).add_(u.to(p.dtype), alpha=-g["lr"])


def build_optimizers(model, lr_muon, lr_adam, wd=0.01):
    mu, ad = [], []
    for n, p in model.named_parameters():
        (mu if p.ndim == 2 and "embed" not in n else ad).append(p)
    return [Muon(mu, lr=lr_muon, wd=wd), torch.optim.AdamW(ad, lr=lr_adam, betas=(0.9, 0.95), weight_decay=0.0)]


def lr_mult(prog, warm=0.03):
    if prog < warm:
        return max(prog / warm, 0.02)
    return 0.1 + 0.9 * 0.5 * (1 + math.cos(math.pi * min(1.0, (prog - warm) / (1 - warm))))


# ----------------------------------------------------------------------------------------------
# Hugging Face Hub (background thread). Auth: env / Colab secret / cached login / interactive login()
# ----------------------------------------------------------------------------------------------
def hf_api():
    from huggingface_hub import HfApi, login
    tok = os.environ.get("HF_TOKEN") or os.environ.get("HUGGINGFACE_HUB_TOKEN")
    if not tok:
        try:
            from google.colab import userdata
            tok = userdata.get("HF_TOKEN")
        except Exception:
            tok = None
    if tok:
        login(token=tok, add_to_git_credential=False)
    api = HfApi()
    try:
        api.whoami()
    except Exception:
        login()
    return api


def resolve_path(path_or_repo, local_dir=None):
    if os.path.isdir(path_or_repo):
        return path_or_repo
    from huggingface_hub import snapshot_download
    hf_api()
    return snapshot_download(repo_id=path_or_repo, local_dir=local_dir or os.path.join("hub_cache", path_or_repo.replace("/", "__")))


class HubSync:
    def __init__(self, repo_id):
        self.repo_id, self.api = repo_id, hf_api()
        self.api.create_repo(repo_id, exist_ok=True, private=True)
        self.q, self.busy = queue.Queue(maxsize=1), threading.Event()
        threading.Thread(target=self._loop, daemon=True).start()

    def _loop(self):
        while True:
            folder, msg = self.q.get()
            self.busy.set()
            try:
                self.api.upload_folder(repo_id=self.repo_id, folder_path=folder, commit_message=msg)
                print(f"[hub] uploaded: {msg}", flush=True)
            except Exception as e:
                print(f"[hub] upload failed ({e}); will retry at the next checkpoint", flush=True)
            finally:
                self.busy.clear()
                self.q.task_done()

    def push(self, folder, msg):
        if self.busy.is_set() or self.q.full():
            return False
        self.q.put_nowait((folder, msg))
        return True

    def flush(self):
        self.q.join()


def save_bundle(d, model, tok):
    from safetensors.torch import save_file
    os.makedirs(d, exist_ok=True)
    save_file({k: v.detach().cpu().contiguous() for k, v in model.state_dict().items()}, os.path.join(d, "model.safetensors"), metadata={"format": "pt"})
    cfg = model.config
    cfg.auto_map = {"AutoConfig": f"{MODULE_NAME}.BarameterConfig", "AutoModelForCausalLM": f"{MODULE_NAME}.BarameterForCausalLM"}
    cfg.save_pretrained(d)
    tok.save(d)
    model.mem.save(os.path.join(d, "memory"))
    src = globals().get("__file__")
    if src and os.path.exists(src):
        shutil.copy(src, os.path.join(d, MODULE_NAME + ".py"))


def load_bundle(path_or_repo, device, mmap=False):
    from safetensors.torch import load_file
    d = resolve_path(path_or_repo)
    cfg = BarameterConfig.from_pretrained(d)
    model = BarameterForCausalLM(cfg)
    model.load_state_dict(load_file(os.path.join(d, "model.safetensors")), strict=True)
    model.mem.load(os.path.join(d, "memory"), mmap)
    return model.to(device), Tok.load(d), d


# ----------------------------------------------------------------------------------------------
# Training (one continuous loop: gradient steps + barameter writes from the same forward pass)
# ----------------------------------------------------------------------------------------------
def pick_device():
    return torch.device("cuda" if torch.cuda.is_available() else "cpu")


def write_batch(model, ids, tag, writes=None):
    writes = writes or model.last_writes
    if writes is None:
        return
    C = model.config.write_chunk
    nc = writes["keys"].shape[1]
    tokens = ids[:, : nc * C].reshape(-1, C).cpu().numpy()
    model.mem.write(writes["keys"].detach().reshape(-1, model.config.d_k).float().cpu(),
                    writes["codes"].detach().reshape(-1, model.config.pq_m).cpu().numpy(), tokens, tag)


@torch.no_grad()
def evaluate(model, tok, texts, device, L, B=8, max_windows=64):
    model.eval()
    wins = []
    for w in _pack(tok, texts, L):
        wins.append(w)
        if len(wins) >= max_windows:
            break
    if not wins:
        return float("nan"), float("nan")
    res = {True: [], False: []}
    for use in (True, False):
        for i in range(0, len(wins), B):
            ids = torch.from_numpy(np.stack(wins[i:i + B])).to(device)
            res[use].append(model(ids, labels=ids, use_store=use, reduction="none").loss.cpu())
    on, off = torch.cat(res[True]).mean().item(), torch.cat(res[False]).mean().item()
    return on, off


def train(model, tok, make_rows, held, a, device):
    os.makedirs(a.out, exist_ok=True)
    model.to(device).train()
    opts = build_optimizers(model, a.lr_muon, a.lr_adam)
    base_lrs = [[g["lr"] for g in o.param_groups] for o in opts]
    amp = device.type == "cuda"
    scaler = torch.amp.GradScaler("cuda", enabled=amp)
    state = {"step": 0, "windows": 0, "elapsed": 0.0}
    sp = os.path.join(a.out, "trainer_state.pt")
    if a.resume and os.path.exists(sp):
        st = torch.load(sp, map_location="cpu", weights_only=False)
        for o, s in zip(opts, st["opts"]):
            o.load_state_dict(s)
        scaler.load_state_dict(st["scaler"])
        torch.set_rng_state(st["rng"])
        state = st["state"]
        print(f"[resume] step {state['step']} windows {state['windows']} elapsed {state['elapsed']:.0f}s", flush=True)
    hub = HubSync(a.hub_repo) if a.hub_repo else None
    stream = window_stream(tok, make_rows, a.seq_len, skip=state["windows"])
    t_start, last_ck, tok_count, t_log = time.time() - state["elapsed"], time.time(), 0, time.time()
    stage = os.path.join(a.out, "hub_stage")

    def checkpoint_now(final=False):
        state["elapsed"] = time.time() - t_start
        if hub and not final and (hub.busy.is_set() or hub.q.full()):
            return
        save_bundle(stage, model, tok)
        torch.save({"opts": [o.state_dict() for o in opts], "scaler": scaler.state_dict(), "rng": torch.get_rng_state(),
                    "state": dict(state)}, os.path.join(stage, "trainer_state.pt"))
        shutil.copy(os.path.join(stage, "trainer_state.pt"), sp)
        if hub:
            if final:
                hub.flush()
                hub.push(stage, f"final step {state['step']}")
                hub.flush()
            else:
                hub.push(stage, f"checkpoint step {state['step']}")

    while True:
        elapsed = time.time() - t_start
        prog = elapsed / (a.minutes * 60) if a.minutes else state["step"] / max(1, a.max_steps)
        if prog >= 1 or (a.max_steps and state["step"] >= a.max_steps):
            break
        try:
            batch = np.stack([next(stream) for _ in range(a.batch)])
        except StopIteration:
            break
        ids = torch.from_numpy(batch).to(device)
        m = lr_mult(prog)
        for o, bl in zip(opts, base_lrs):
            for g, b in zip(o.param_groups, bl):
                g["lr"] = b * m
        with torch.autocast(device_type=device.type, dtype=torch.float16, enabled=amp):
            out = model(ids, labels=ids)
        scaler.scale(out.loss).backward()
        for o in opts:
            scaler.unscale_(o)
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        for o in opts:
            scaler.step(o)
        scaler.update()
        for o in opts:
            o.zero_grad(set_to_none=True)
        with torch.no_grad():
            write_batch(model, ids, a.tag)
        state["step"] += 1
        state["windows"] += a.batch
        tok_count += ids.numel()
        if state["step"] % a.dead_every == 0:
            dead = model.writer.reinit_dead()
            if dead:
                print(f"[codebook] reinitialised {dead} dead codes", flush=True)
        if state["step"] % a.log_every == 0:
            aux = {k: round(float(v), 4) for k, v in model.last_aux.items()}
            print(f"step {state['step']:5d} loss {float(out.loss.detach()):.4f} {aux} lr x{m:.2f} "
                  f"{tok_count / (time.time() - t_log):.0f} tok/s store {model.mem.store.n} index {model.mem.index.n} "
                  f"elapsed {time.time() - t_start:.0f}s", flush=True)
            tok_count, t_log = 0, time.time()
        if a.ckpt_minutes and time.time() - last_ck > a.ckpt_minutes * 60:
            checkpoint_now()
            last_ck = time.time()
    checkpoint_now(final=True)
    if held:
        on, off = evaluate(model, tok, held, device, a.seq_len)
        print(f"[heldout] loss with barameters {on:.4f} | store disabled {off:.4f}", flush=True)
    model.train()
    return state


# ----------------------------------------------------------------------------------------------
# Search / answer / reindex / forgetting scan / refresh
# ----------------------------------------------------------------------------------------------
@torch.no_grad()
def window_ids(mem, w, per, device):
    return torch.from_numpy(mem.index.window(w, per).astype(np.int64))[None].to(device)


@torch.no_grad()
def reencode_windows(model, windows, device, per, tag_ds=None):
    """Re-encode index windows with the current writer; refreshes index keys and existing store slots."""
    mem, C = model.mem, model.config.write_chunk
    for w in windows:
        ids = window_ids(mem, w, per, device)
        wr = model.encode_chunks(ids)
        if wr is None:
            continue
        rows = np.arange(w * per, w * per + wr["keys"].shape[1])
        k = wr["keys"][0].float().cpu().numpy().astype(np.float16)
        mem.index.keys.buf[rows] = k
        mem.index.invalidate()
        slots = np.nonzero(np.isin(mem.store.chunk.data, rows))[0]
        if len(slots):
            pos = {int(r): i for i, r in enumerate(rows)}
            for s in slots:
                i = pos[int(mem.store.chunk.data[s])]
                mem.store.keys.buf[s] = k[i]
                mem.store.codes.buf[s] = wr["codes"][0, i].cpu().numpy().astype(np.uint8)
    mem.store.invalidate()


@torch.no_grad()
def reindex(model, device, seq_len):
    per = seq_len // model.config.write_chunk
    nw = math.ceil(model.mem.index.n / per)
    reencode_windows(model, range(nw), device, per)
    return nw


@torch.no_grad()
def deep_search(model, tok, text, device, topm=3):
    """Encode the query as one chunk, brute-force search the full ChunkIndex. Returns [(score,row,text)]."""
    mem = model.mem
    ids = torch.tensor([tok.encode(text)], device=device)
    wr = model.encode_chunks(ids, chunk=ids.shape[1])
    if wr is None or mem.index.n == 0:
        return []
    sc, ix = mem.index.search(wr["keys"][0].float(), topm)
    return [(float(s), int(i), tok.decode(mem.index.tok.data[int(i)])) for s, i in zip(sc[0], ix[0])]


@torch.no_grad()
def answer(model, tok, question, device, deep=False, score_thr=0.5, ent_thr=4.0, topm=3, expand=2, max_new=128, temperature=0.7):
    model.eval()
    mem = model.mem
    prompt = tok.encode(f"<|user|>\n{question}\n<|assistant|>\n")
    pid = torch.tensor([prompt], device=device)
    info = {"deep": False, "best_score": None, "entropy": None, "hits": []}
    p = torch.softmax(model(pid).logits[0, -1].float(), -1)
    info["entropy"] = float(-(p * torch.log(p + 1e-9)).sum())
    wr = model.encode_chunks(pid, chunk=pid.shape[1])
    if wr is not None and mem.store.n:
        info["best_score"] = float(mem.store.search(wr["keys"][0].float(), 1)[0][0, 0])
    if deep or info["best_score"] is None or info["best_score"] < score_thr or info["entropy"] > ent_thr:
        hits = deep_search(model, tok, question, device, topm)
        info["deep"], info["hits"] = True, hits
        ctx = []
        for _, row, _ in hits:
            ctx += [int(t) for r in range(row, min(row + 1 + expand, mem.index.n)) for t in mem.index.tok.data[r]]
        prompt = (ctx + prompt)[-(model.config.max_seq_len - max_new):]
        pid = torch.tensor([prompt], device=device)
    out = model.chat_generate(pid, max_new=max_new, temperature=temperature)[0, len(prompt):]
    return tok.decode(out).replace("<eos>", "").strip(), info


@torch.no_grad()
def forgetting_scan(model, device, seq_len, n_windows=128, B=8):
    mem = model.mem
    per = seq_len // model.config.write_chunk
    nw = mem.index.n // per
    if nw == 0:
        return []
    model.eval()
    pick = random.sample(range(nw), min(n_windows, nw))
    res = []
    for i in range(0, len(pick), B):
        ws = pick[i:i + B]
        ids = torch.cat([window_ids(mem, w, per, device) for w in ws])
        off = model(ids, labels=ids, use_store=False, reduction="none").loss.cpu()
        on = model(ids, labels=ids, use_store=True, reduction="none").loss.cpu()
        res += [(float(o), float(n), w) for o, n, w in zip(off, on, ws)]
    return sorted(res, reverse=True)  # worst (store-off loss highest) first


def refresh(model, device, seq_len, steps=20, worst=32, replay=32, lr=3e-5, B=8, scan=128):
    mem, per = model.mem, seq_len // model.config.write_chunk
    scanned = forgetting_scan(model, device, seq_len, scan, B)
    if not scanned:
        return {"scanned": 0}
    bad = [w for _, _, w in scanned[:worst]]
    rest = [w for _, _, w in scanned[worst:]] or bad
    before = sum(s[0] for s in scanned[:worst]) / len(bad)
    opt = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=0.0)
    model.train()
    for _ in range(steps):  # half worst, half replay so repairing one thing does not erase another
        ws = random.sample(bad, min(B // 2, len(bad))) + random.sample(rest, min(B - B // 2, len(rest)))
        ids = torch.cat([window_ids(mem, w, per, device) for w in ws])
        loss = model(ids, labels=ids, use_store=False).loss
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        opt.step(); opt.zero_grad(set_to_none=True)
    model.eval()
    reencode_windows(model, bad, device, per)  # rewrite barameters for the repaired chunks (fixes key drift too)
    with torch.no_grad():
        after = sum(float(model(torch.cat([window_ids(mem, w, per, device) for w in bad[i:i + B]]), labels=torch.cat(
            [window_ids(mem, w, per, device) for w in bad[i:i + B]]), use_store=False, reduction="none").loss.sum())
                    for i in range(0, len(bad), B)) / len(bad)
    return {"scanned": len(scanned), "repaired": len(bad), "loss_before": before, "loss_after": after}


# ----------------------------------------------------------------------------------------------
# Self-test (CPU, no network)
# ----------------------------------------------------------------------------------------------
def selftest():
    torch.manual_seed(0); random.seed(0); np.random.seed(0)
    dev = torch.device("cpu")
    ok = lambda name: print(f"  PASS  {name}", flush=True)
    # 0. default-size parameter count
    big = BarameterForCausalLM(BarameterConfig())
    print(f"  default config params: {big.num_params() / 1e6:.1f}M (vocab 16k)")
    del big
    # 1. recurrence correctness
    q, k = F.normalize(torch.randn(2, 37, 3, 16), dim=-1), F.normalize(torch.randn(2, 37, 3, 16), dim=-1)
    v, g = torch.randn(2, 37, 3, 16), -F.softplus(torch.randn(2, 37, 3))
    err = (chunked_gated_linear_attention(q, k, v, g, 8) - reference_recurrence(q, k, v, g)).abs().max().item()
    assert err < 1e-4, err
    ok(f"chunked recurrence == sequential reference (max err {err:.2e})")
    # data
    caps = {f"Land{i}": f"City{(i * 7) % 50}" for i in range(120)}
    rows = [f"<|user|>\nWhat is the capital of {c}?\n<|assistant|>\nThe capital of {c} is {v}." for c, v in caps.items()]
    held, trainrows = rows[:10], rows[10:]
    tok = Tok.train(rows, 400)
    cfg = BarameterConfig(vocab_size=tok.vocab_size, d_model=128, n_heads=4, n_kv_heads=2, head_dim=32, ffn_hidden=192,
                          block_types=["lin", "lin", "attn", "lin", "lin", "attn"], read_after=[2, 4], write_from=3,
                          rec_chunk=16, write_chunk=8, max_seq_len=64, store_slots=300, d_k=32, d_v=64, pq_m=4, pq_k=16, topk=4)
    model = BarameterForCausalLM(cfg)
    ids = torch.randint(4, cfg.vocab_size, (2, 64))
    model.train()
    out = model(ids, labels=ids)
    out.loss.backward()
    assert torch.isfinite(out.loss) and all(torch.isfinite(p.grad).all() for p in model.parameters() if p.grad is not None)
    ok(f"forward/backward finite (loss {out.loss.item():.3f}, ln V = {math.log(cfg.vocab_size):.3f}, aux {sorted(model.last_aux)})")
    model.zero_grad()
    # 2. training loop with barameter writes
    tmp = tempfile.mkdtemp()
    a = argparse.Namespace(out=tmp, hub_repo=None, minutes=0, max_steps=80, batch=8, seq_len=64, lr_muon=0.02, lr_adam=2e-3,
                           ckpt_minutes=0, tag="capitals", log_every=20, dead_every=40, resume=False)
    make_rows = lambda ep: iter(random.Random(ep).sample(trainrows, len(trainrows)))
    st = train(model, tok, make_rows, held, a, dev)
    assert st["step"] == 80 and model.mem.store.n > 0 and model.mem.index.n > 0 and model.mem.store.n <= cfg.store_slots
    ok(f"train loop: store {model.mem.store.n}/{cfg.store_slots} slots, index {model.mem.index.n} chunks")
    # 3. generate + deep search + answer
    s_, i_ = answer(model, tok, "What is the capital of Land3?", dev, deep=True, max_new=12)
    assert i_["deep"] and len(i_["hits"]) > 0
    ok(f"answer() with deep search ran ({len(i_['hits'])} hits)")
    # 4. forgetting scan + refresh
    r = refresh(model, dev, 64, steps=3, worst=4, scan=8, B=4)
    assert r["scanned"] > 0
    ok(f"refresh ran: {r}")
    nw = reindex(model, dev, 64)
    ok(f"reindex re-encoded {nw} windows")
    # 5. growth is function-preserving
    model.eval()
    base = model(ids).logits
    grown = grow_model(model, add_layers=2, new_d_model=256).eval()
    diff = (grown(ids).logits - base).abs().max().item()
    assert diff < 1e-3, diff
    ok(f"grow {model.num_params() / 1e6:.2f}M -> {grown.num_params() / 1e6:.2f}M params ({cfg.n_layers}->{grown.config.n_layers} layers, "
       f"d {cfg.d_model}->{grown.config.d_model}); max logit diff {diff:.2e}")
    gl = grown.train()(ids, labels=ids)
    gl.loss.backward()
    ok("grown model trains (backward finite)")
    # 6. save / load roundtrip
    save_bundle(tmp + "/ck", grown.eval(), tok)
    m2, t2, _ = load_bundle(tmp + "/ck", dev)
    assert (m2.eval()(ids).logits - grown(ids).logits).abs().max().item() < 1e-5 and m2.mem.store.n == grown.mem.store.n
    ok("save_bundle/load_bundle roundtrip (weights + store + index)")
    try:
        m3 = AutoModelForCausalLM.from_pretrained(tmp + "/ck")
        ok("AutoModelForCausalLM.from_pretrained on the saved folder")
    except Exception as e:
        print(f"  WARN  AutoModelForCausalLM.from_pretrained failed: {e}")
    m4 = AutoModelForCausalLM.from_config(cfg)
    assert isinstance(m4, BarameterForCausalLM)
    ok("AutoClass registration (AutoModelForCausalLM.from_config)")
    # 7. delete a dataset -> forgotten
    n0 = model.mem.store.n
    ns, ni = model.mem.delete("capitals")
    assert model.mem.store.n == 0 and model.mem.index.n == 0 and ns == n0
    ok(f"delete dataset tag: removed {ns} barameters and {ni} chunks")
    print("ALL SELF-TESTS PASSED")


# ----------------------------------------------------------------------------------------------
# CLI
# ----------------------------------------------------------------------------------------------
def main(argv=None):
    ap = argparse.ArgumentParser(description="BarameterLM")
    sub = ap.add_subparsers(dest="cmd", required=True)

    def common(p, ck=True):
        if ck:
            p.add_argument("--ckpt", default="brucoder/BarameterLM")
        p.add_argument("--mmap", action="store_true")
        p.add_argument("--push", default=None, help="repo id to push the updated bundle to")

    sub.add_parser("selftest")
    t = sub.add_parser("train"); common(t, False)
    t.add_argument("--dataset", default="yahma/alpaca-cleaned"); t.add_argument("--split", default="train")
    t.add_argument("--streaming", action="store_true"); t.add_argument("--max_examples", type=int, default=0)
    t.add_argument("--holdout", type=int, default=500); t.add_argument("--tag", default=None)
    t.add_argument("--out", default="barameter_run"); t.add_argument("--hub_repo", default="brucoder/BarameterLM")
    t.add_argument("--minutes", type=float, default=10); t.add_argument("--max_steps", type=int, default=0)
    t.add_argument("--batch", type=int, default=8); t.add_argument("--seq_len", type=int, default=512)
    t.add_argument("--vocab", type=int, default=16384); t.add_argument("--lr_muon", type=float, default=0.02)
    t.add_argument("--lr_adam", type=float, default=1e-3); t.add_argument("--ckpt_minutes", type=float, default=5)
    t.add_argument("--log_every", type=int, default=20); t.add_argument("--dead_every", type=int, default=200)
    t.add_argument("--resume", action="store_true"); t.add_argument("--grad_checkpoint", action="store_true")
    t.add_argument("--seed", type=int, default=0)
    ig = sub.add_parser("ingest"); common(ig)
    ig.add_argument("--dataset", required=True); ig.add_argument("--tag", required=True); ig.add_argument("--split", default="train")
    ig.add_argument("--streaming", action="store_true"); ig.add_argument("--max_examples", type=int, default=0)
    ig.add_argument("--seq_len", type=int, default=512); ig.add_argument("--batch", type=int, default=16)
    for name in ("ask", "search"):
        p = sub.add_parser(name); common(p); p.add_argument("question"); p.add_argument("--deep", action="store_true")
        p.add_argument("--max_new", type=int, default=128); p.add_argument("--temperature", type=float, default=0.7)
    ch = sub.add_parser("chat"); common(ch); ch.add_argument("--deep", action="store_true")
    rf = sub.add_parser("refresh"); common(rf)
    rf.add_argument("--steps", type=int, default=20); rf.add_argument("--worst", type=int, default=32)
    rf.add_argument("--scan", type=int, default=128); rf.add_argument("--seq_len", type=int, default=512)
    rf.add_argument("--lr", type=float, default=3e-5)
    ri = sub.add_parser("reindex"); common(ri); ri.add_argument("--seq_len", type=int, default=512)
    ev = sub.add_parser("eval"); common(ev)
    ev.add_argument("--dataset", default="yahma/alpaca-cleaned"); ev.add_argument("--seq_len", type=int, default=512)
    ev.add_argument("--holdout", type=int, default=500)
    sub.add_parser("list").add_argument("--ckpt", default="brucoder/BarameterLM")
    ins = sub.add_parser("inspect"); ins.add_argument("--ckpt", default="brucoder/BarameterLM"); ins.add_argument("slot", type=int)
    de = sub.add_parser("delete"); common(de); de.add_argument("tag")
    gr = sub.add_parser("grow"); common(gr)
    gr.add_argument("--add_layers", type=int, default=0); gr.add_argument("--d_model", type=int, default=0)
    gr.add_argument("--out", default="barameter_grown")
    args = ap.parse_args(argv)
    dev = pick_device()

    if args.cmd == "selftest":
        return selftest()

    if args.cmd == "train":
        torch.manual_seed(args.seed); random.seed(args.seed)
        args.tag = args.tag or args.dataset.split("/")[-1]
        make_rows, held = load_rows(args.dataset, args.split, args.streaming, args.max_examples, args.holdout)
        resume_dir = os.path.join(args.out, "hub_stage")
        if args.resume and not os.path.exists(os.path.join(args.out, "trainer_state.pt")) and args.hub_repo:
            resolve_path(args.hub_repo, resume_dir)
            shutil.copy(os.path.join(resume_dir, "trainer_state.pt"), os.path.join(args.out, "trainer_state.pt")) if os.path.exists(os.path.join(resume_dir, "trainer_state.pt")) else None
        if args.resume and os.path.exists(os.path.join(resume_dir, "model.safetensors")):
            model, tok, _ = load_bundle(resume_dir, dev)
        else:
            sample = []
            for r in make_rows(0):
                sample.append(r)
                if len(sample) >= 20000:
                    break
            tok = Tok.train(sample, args.vocab)
            cfg = BarameterConfig(vocab_size=tok.vocab_size, max_seq_len=args.seq_len, grad_checkpoint=args.grad_checkpoint)
            model = BarameterForCausalLM(cfg)
        print(f"model: {model.num_params() / 1e6:.1f}M parameters, device {dev}, tag '{args.tag}'", flush=True)
        train(model, tok, make_rows, held, args, dev)
        return

    if args.cmd in ("list", "inspect"):
        model, tok, d = load_bundle(args.ckpt, torch.device("cpu"))
        m = model.mem
        if args.cmd == "list":
            print(f"params {model.num_params() / 1e6:.1f}M | store {m.store.n}/{m.store.cap} slots | index {m.index.n} chunks | growth {model.config.growth_history}")
            for tag, (ns, ni) in m.counts().items():
                print(f"  {tag:30s} barameters {ns:8d}  chunks {ni:8d}")
        else:
            s = args.slot
            c = int(m.store.chunk.data[s])
            print(f"slot {s} dataset={m.datasets[int(m.store.ds.data[s])]} usage={int(m.store.usage.data[s])} "
                  f"step={int(m.store.step.data[s])} codes={m.store.codes.data[s].tolist()} chunk_row={c}")
            if c >= 0:
                print("text:", repr(tok.decode(m.index.tok.data[c])))
        return

    model, tok, d = load_bundle(args.ckpt, dev, getattr(args, "mmap", False))
    model.eval()
    dirty = False
    if args.cmd in ("ask", "chat"):
        qs = [args.question] if args.cmd == "ask" else iter(lambda: input("you> "), None)
        for q in qs:
            ans, info = answer(model, tok, q, dev, deep=args.deep, max_new=getattr(args, "max_new", 128), temperature=getattr(args, "temperature", 0.7))
            print(f"[best barameter score {info['best_score']} | entropy {info['entropy']:.2f} | deep search {'yes' if info['deep'] else 'no'}]")
            print(ans)
    elif args.cmd == "search":
        for s, row, text in deep_search(model, tok, args.question, dev, 5):
            print(f"{s:.3f}  row {row}  {text!r}")
    elif args.cmd == "ingest":
        make_rows, _ = load_rows(args.dataset, args.split, args.streaming, args.max_examples, 0)
        wins, n = [], 0
        for w in window_stream(tok, lambda ep: make_rows(0) if ep == 0 else iter(()), args.seq_len):
            wins.append(w)
            if len(wins) == args.batch:
                ids = torch.from_numpy(np.stack(wins)).to(dev)
                write_batch(model, ids, args.tag, model.encode_chunks(ids))
                n += len(wins); wins = []
                if (n // args.batch) % 20 == 0:
                    print(f"ingested {n} windows; store {model.mem.store.n}, index {model.mem.index.n}", flush=True)
        if wins:
            ids = torch.from_numpy(np.stack(wins)).to(dev)
            write_batch(model, ids, args.tag, model.encode_chunks(ids))
        dirty = True
    elif args.cmd == "refresh":
        print(refresh(model, dev, args.seq_len, args.steps, args.worst, lr=args.lr, scan=args.scan)); dirty = True
    elif args.cmd == "reindex":
        print(f"re-encoded {reindex(model, dev, args.seq_len)} windows"); dirty = True
    elif args.cmd == "eval":
        _, held = load_rows(args.dataset, "train", False, 1, args.holdout)
        on, off = evaluate(model, tok, held, dev, args.seq_len)
        print(f"held-out loss: with barameters {on:.4f} | store disabled {off:.4f}")
    elif args.cmd == "delete":
        print("removed (barameters, chunks):", model.mem.delete(args.tag)); dirty = True
    elif args.cmd == "grow":
        grown = grow_model(model, args.add_layers, args.d_model or None)
        print(f"grown to {grown.num_params() / 1e6:.1f}M parameters, history {grown.config.growth_history}")
        save_bundle(args.out, grown, tok)
        if args.push:
            api = hf_api(); api.create_repo(args.push, exist_ok=True, private=True)
            api.upload_folder(repo_id=args.push, folder_path=args.out)
        return
    if dirty:
        save_bundle(d, model, tok)
        if args.push:
            api = hf_api(); api.create_repo(args.push, exist_ok=True, private=True)
            api.upload_folder(repo_id=args.push, folder_path=d)


if __name__ == "__main__" and "ipykernel" not in sys.modules:
    main()
```

```python
#!/usr/bin/env python
"""
BarameterLM - one file: config, model, barameter memory, training, ingestion, search, refresh, growth, CLI.

  python barameter.py selftest                                   # CPU self-test (no network)
  python barameter.py train --dataset yahma/alpaca-cleaned --minutes 10 --hub_repo brucoder/BarameterLM
  python barameter.py train --resume --hub_repo brucoder/BarameterLM --minutes 10
  python barameter.py ingest  --ckpt brucoder/BarameterLM --dataset databricks/databricks-dolly-15k --tag dolly
  python barameter.py ask     --ckpt brucoder/BarameterLM "What is the capital of France?" [--deep]
  python barameter.py chat    --ckpt brucoder/BarameterLM
  python barameter.py search  --ckpt brucoder/BarameterLM "query"       # deep GPU search over every stored chunk
  python barameter.py refresh --ckpt brucoder/BarameterLM --steps 20    # find forgotten chunks, rewrite + short retrain
  python barameter.py reindex --ckpt brucoder/BarameterLM               # re-encode all keys with the current writer
  python barameter.py eval    --ckpt brucoder/BarameterLM --dataset yahma/alpaca-cleaned
  python barameter.py list | inspect <slot> | delete <tag>  --ckpt ...
  python barameter.py grow    --ckpt ... --add_layers 12 --d_model 1280 --out grown [--push brucoder/BarameterLM-300M]

Memory design
  hot tier  : BarameterStore  - fixed slots, L2-normalised 128-d keys + 4x8-bit product-quantised values (4 bytes/slot),
              evicts least-used when full, merges near-duplicate keys, every entry tagged with a dataset.
  cold tier : ChunkIndex      - EVERY chunk ever seen (key + tokens), brute-force GPU search, used by the deep path.
  Search is chunked over slots and arrays are numpy-backed (optionally np.memmap, --mmap), so the store can grow
  from 65k slots in VRAM to millions of disk-mapped slots without touching the backbone.
Auth: never hard-coded. Uses HF_TOKEN env / Colab secret HF_TOKEN / cached login, else an interactive login().
"""
import os, sys, json, math, time, copy, queue, shutil, hashlib, argparse, random, threading, tempfile
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.checkpoint import checkpoint
from transformers import PretrainedConfig, PreTrainedModel, AutoConfig, AutoModelForCausalLM
from transformers.modeling_outputs import CausalLMOutput

MODULE_NAME = os.path.splitext(os.path.basename(globals().get("__file__", "barameter.py")))[0]
SPECIALS = ["<pad>", "<eos>", "<|user|>", "<|assistant|>"]


# ----------------------------------------------------------------------------------------------
# Config
# ----------------------------------------------------------------------------------------------
class BarameterConfig(PretrainedConfig):
    model_type = "barameter"

    def __init__(self, vocab_size=16384, d_model=640, n_heads=10, n_kv_heads=2, head_dim=64, ffn_hidden=1728,
                 block_types=None, read_after=(4, 8), write_from=6, rec_chunk=64, write_chunk=32, max_seq_len=1024,
                 store_slots=65536, d_k=128, d_v=256, pq_m=4, pq_k=256, topk=8, dedup_threshold=0.97,
                 softcap=30.0, rope_theta=10000.0, local_memory=True, grad_checkpoint=False,
                 growth_history=None, **kw):
        if block_types is None:
            block_types = ["attn" if i in (3, 7, 11) else "lin" for i in range(12)]
        assert d_model == n_heads * head_dim, "d_model must equal n_heads*head_dim"
        assert n_heads % n_kv_heads == 0 and d_v % pq_m == 0 and pq_k <= 256
        self.vocab_size, self.d_model, self.n_heads, self.n_kv_heads, self.head_dim = vocab_size, d_model, n_heads, n_kv_heads, head_dim
        self.ffn_hidden, self.block_types = ffn_hidden, list(block_types)
        self.read_after, self.write_from = [int(x) for x in read_after], int(write_from)
        self.rec_chunk, self.write_chunk, self.max_seq_len = rec_chunk, write_chunk, max_seq_len
        self.store_slots, self.d_k, self.d_v, self.pq_m, self.pq_k, self.topk = store_slots, d_k, d_v, pq_m, pq_k, topk
        self.dedup_threshold, self.softcap, self.rope_theta = dedup_threshold, softcap, rope_theta
        self.local_memory, self.grad_checkpoint = local_memory, grad_checkpoint
        self.growth_history = list(growth_history or [])
        self.n_layers = len(self.block_types)
        kw["tie_word_embeddings"] = False  # tying is done manually (embedding matrix is reused as the LM head)
        kw.setdefault("pad_token_id", 0)
        kw.setdefault("eos_token_id", 1)
        super().__init__(**kw)


# ----------------------------------------------------------------------------------------------
# Backbone
# ----------------------------------------------------------------------------------------------
class RMSNorm(nn.Module):
    def __init__(self, d, eps=1e-8):
        super().__init__()
        self.weight, self.eps = nn.Parameter(torch.ones(d)), eps

    def forward(self, x):
        xf = x.float()
        return (xf * torch.rsqrt(xf.pow(2).mean(-1, keepdim=True) + self.eps)).to(x.dtype) * self.weight


def rope(x, theta):  # x: (B,H,L,Dh)
    half = x.shape[-1] // 2
    freqs = 1.0 / (theta ** (torch.arange(half, device=x.device, dtype=torch.float32) / half))
    ang = torch.arange(x.shape[2], device=x.device, dtype=torch.float32)[:, None] * freqs[None]
    cos, sin = ang.cos(), ang.sin()
    x1, x2 = x[..., :half].float(), x[..., half:].float()
    return torch.cat([x1 * cos - x2 * sin, x1 * sin + x2 * cos], -1).to(x.dtype)


class SwiGLU(nn.Module):
    def __init__(self, d, h):
        super().__init__()
        self.w1, self.w3, self.w2 = nn.Linear(d, h, bias=False), nn.Linear(d, h, bias=False), nn.Linear(h, d, bias=False)

    def forward(self, x):
        return self.w2(F.silu(self.w1(x)) * self.w3(x))


def chunked_gated_linear_attention(q, k, v, g, chunk):
    """q,k,v: (B,L,H,dh) float32 (q,k L2-normalised); g: (B,L,H) log-decay <= 0.  Returns (B,L,H,dh).
    Within a chunk: decay-masked matmul.  Between chunks: carried state S (B,H,dh,dh).  All in fp32, log-space decay."""
    B, L, H, dh = q.shape
    pad = (-L) % chunk
    if pad:
        q, k, v = [F.pad(t, (0, 0, 0, 0, 0, pad)) for t in (q, k, v)]
        g = F.pad(g, (0, 0, 0, pad))
    Lp = L + pad
    nc = Lp // chunk
    q, k, v = [t.reshape(B, nc, chunk, H, dh).permute(0, 3, 1, 2, 4) for t in (q, k, v)]  # B,H,nc,C,dh
    g = g.reshape(B, nc, chunk, H).permute(0, 3, 1, 2)  # B,H,nc,C
    b = g.cumsum(-1)
    mask = torch.tril(torch.ones(chunk, chunk, dtype=torch.bool, device=q.device))
    diff = (b.unsqueeze(-1) - b.unsqueeze(-2)).masked_fill(~mask, float("-inf"))
    intra = torch.matmul(torch.matmul(q, k.transpose(-1, -2)) * torch.exp(diff), v)
    qd = q * torch.exp(b).unsqueeze(-1)
    kd = k * torch.exp(b[..., -1:] - b).unsqueeze(-1)
    total = torch.exp(b[..., -1])
    S = q.new_zeros(B, H, dh, dh)
    outs = []
    for c in range(nc):
        outs.append(torch.matmul(qd[:, :, c], S))
        S = S * total[:, :, c, None, None] + torch.matmul(kd[:, :, c].transpose(-1, -2), v[:, :, c])
    o = intra + torch.stack(outs, 2)  # B,H,nc,C,dh
    return o.permute(0, 2, 3, 1, 4).reshape(B, Lp, H, dh)[:, :L]


def reference_recurrence(q, k, v, g):
    B, L, H, dh = q.shape
    S = q.new_zeros(B, H, dh, dh)
    out = []
    for t in range(L):
        S = torch.exp(g[:, t])[..., None, None] * S + k[:, t].unsqueeze(-1) * v[:, t].unsqueeze(-2)
        out.append(torch.einsum("bhd,bhde->bhe", q[:, t], S))
    return torch.stack(out, 1)


class LinearRecurrentMix(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        D, H, dh = cfg.d_model, cfg.n_heads, cfg.head_dim
        self.H, self.dh, self.chunk = H, dh, cfg.rec_chunk
        self.q, self.k, self.v = (nn.Linear(D, H * dh, bias=False) for _ in range(3))
        self.g = nn.Linear(D, H)
        self.o_gate = nn.Linear(D, H * dh, bias=False)
        self.onorm = RMSNorm(dh)
        self.out = nn.Linear(H * dh, D, bias=False)

    def forward(self, x):
        B, L, _ = x.shape
        H, dh = self.H, self.dh
        og = torch.sigmoid(self.o_gate(x)).view(B, L, H, dh)
        with torch.autocast(device_type=x.device.type, enabled=False):
            q = F.normalize(self.q(x).float().view(B, L, H, dh), dim=-1)
            k = F.normalize(self.k(x).float().view(B, L, H, dh), dim=-1)
            v = self.v(x).float().view(B, L, H, dh)
            g = -F.softplus(self.g(x).float())
            o = chunked_gated_linear_attention(q, k, v, g, self.chunk)
        o = self.onorm(o.to(x.dtype)) * og
        return self.out(o.reshape(B, L, H * dh))


class CausalAttention(nn.Module):
    def __init__(self, cfg):
        super().__init__()
        D, H, Hkv, dh = cfg.d_model, cfg.n_heads, cfg.n_kv_heads, cfg.head_dim
        self.H, self.Hkv, self.dh, self.theta = H, Hkv, dh, cfg.rope_theta
        self.q = nn.Linear(D, H * dh, bias=False)
        self.k = nn.Linear(D, Hkv * dh, bias=False)
        self.v = nn.Linear(D, Hkv * dh, bias=False)
        self.out = nn.Linear(H * dh, D, bias=False)
        self.qn, self.kn = RMSNorm(dh), RMSNorm(dh)

    def forward(self, x):
        B, L, _ = x.shape
        q = self.qn(self.q(x).view(B, L, self.H, self.dh)).transpose(1, 2)
        k = self.kn(self.k(x).view(B, L, self.Hkv, self.dh)).transpose(1, 2)
        v = self.v(x).view(B, L, self.Hkv, self.dh).transpose(1, 2)
        q, k = rope(q, self.theta), rope(k, self.theta)
        g = self.H // self.Hkv
        k, v = k.repeat_interleave(g, 1), v.repeat_interleave(g, 1)
        o = F.scaled_dot_product_attention(q, k, v, is_causal=True)
        return self.out(o.transpose(1, 2).reshape(B, L, self.H * self.dh))


class Block(nn.Module):
    def __init__(self, cfg, kind):
        super().__init__()
        self.kind = kind
        self.n1, self.n2 = RMSNorm(cfg.d_model), RMSNorm(cfg.d_model)
        self.mix = CausalAttention(cfg) if kind == "attn" else LinearRecurrentMix(cfg)
        self.ffn = SwiGLU(cfg.d_model, cfg.ffn_hidden)

    def forward(self, x):
        x = x + self.mix(self.n1(x))
        return x + self.ffn(self.n2(x))


# ----------------------------------------------------------------------------------------------
# Barameter memory (numpy-backed, scalable)
# ----------------------------------------------------------------------------------------------
class GrowArray:
    """Append-only numpy array with doubling capacity; can be opened as a read-only memmap."""

    def __init__(self, tail, dtype, cap=1024):
        self.tail, self.dtype = tuple(tail), np.dtype(dtype)
        self.buf, self.n, self.ro = np.zeros((cap,) + self.tail, self.dtype), 0, False

    @property
    def data(self):
        return self.buf[: self.n]

    def append(self, arr):
        if self.ro:
            raise RuntimeError("memory opened read-only (--mmap); reload without --mmap to modify it")
        arr = np.asarray(arr, self.dtype)
        m, need = len(arr), self.n + len(arr)
        if need > len(self.buf):
            nb = np.zeros((max(need, 2 * len(self.buf)),) + self.tail, self.dtype)
            nb[: self.n] = self.buf[: self.n]
            self.buf = nb
        st = self.n
        self.buf[st:st + m] = arr
        self.n += m
        return st

    def keep(self, mask):
        d = np.array(self.buf[: self.n][mask])
        self.buf = np.concatenate([d, np.zeros((16,) + self.tail, self.dtype)])
        self.n, self.ro = len(d), False

    def save(self, path):
        np.save(path, np.ascontiguousarray(self.data))

    def load(self, path, mmap=False):
        a = np.load(path, mmap_mode="r" if mmap else None)
        self.buf, self.n, self.ro = (a, len(a), True) if mmap else (np.concatenate([a, np.zeros((16,) + self.tail, self.dtype)]), len(a), False)


def chunked_topk(keys, q, k, step=16384):
    """Brute-force top-k of q (Q,dk) against keys (N,dk) (torch or numpy/memmap), streaming over N."""
    N = len(keys)
    if N == 0:
        return q.new_zeros(len(q), 0), torch.zeros(len(q), 0, dtype=torch.long, device=q.device)
    best_s = best_i = None
    for a in range(0, N, step):
        kc = keys[a:a + step]
        if isinstance(kc, np.ndarray):
            kc = torch.from_numpy(np.ascontiguousarray(kc))
        sc = q @ kc.to(q.device).to(q.dtype).T
        s, i = sc.topk(min(k, sc.shape[1]), dim=1)
        i = i + a
        if best_s is None:
            best_s, best_i = s, i
        else:
            cs, ci = torch.cat([best_s, s], 1), torch.cat([best_i, i], 1)
            best_s, j = cs.topk(min(k, cs.shape[1]), 1)
            best_i = ci.gather(1, j)
    return best_s.float(), best_i


def hash_rows(tokens):
    return np.array([int.from_bytes(hashlib.blake2b(np.ascontiguousarray(r).tobytes(), digest_size=8).digest(), "little", signed=True)
                     for r in tokens], np.int64)


class _KeyedArrays:
    """Shared helpers: GPU key cache and sorted-hash lookup."""
    _cache = None
    _hs = None

    def _qdtype(self, q):
        return q.half() if q.is_cuda else q.float()

    def keys_for(self, device):
        if device.type == "cuda" and self.keys.n * self.keys.tail[0] * 2 < 1.5e9:
            if self._cache is None or len(self._cache) != self.keys.n:
                self._cache = torch.from_numpy(np.ascontiguousarray(self.keys.data)).to(device)
            return self._cache
        return self.keys.data

    def search(self, q, k):
        return chunked_topk(self.keys_for(q.device), self._qdtype(q), k)

    def lookup(self, h):
        if self._hs is None:
            order = np.argsort(self.hash.data, kind="stable")
            self._hs, self._ho = self.hash.data[order], order
        if len(self._hs) == 0:
            return np.zeros(len(h), np.int64), np.zeros(len(h), bool)
        pos = np.minimum(np.searchsorted(self._hs, h), len(self._hs) - 1)
        return self._ho[pos], self._hs[pos] == h

    def invalidate(self, hashes_changed=False):
        self._cache = None
        if hashes_changed:
            self._hs = None


class ChunkIndex(_KeyedArrays):
    """Cold tier: every chunk ever seen (key + tokens). Never evicts; used by the deep search path."""

    def __init__(self, cfg):
        self.keys = GrowArray((cfg.d_k,), np.float16)
        self.tok = GrowArray((cfg.write_chunk,), np.int32)
        self.ds = GrowArray((), np.int32)
        self.hash = GrowArray((), np.int64)

    @property
    def n(self):
        return self.keys.n

    def add(self, keys, tokens, ds):
        tokens = np.ascontiguousarray(tokens, np.int32)
        h, keys = hash_rows(tokens), np.asarray(keys, np.float16)
        slot, hit = self.lookup(h)
        rows = np.zeros(len(h), np.int64)
        rows[hit] = slot[hit]
        self.keys.buf[rows[hit]] = keys[hit]  # refresh with the newest key
        new = ~hit
        if new.any():
            st = self.keys.append(keys[new])
            self.tok.append(tokens[new]); self.ds.append(np.full(int(new.sum()), ds, np.int32)); self.hash.append(h[new])
            rows[new] = np.arange(st, st + int(new.sum()))
            self.invalidate(True)
        else:
            self.invalidate()
        return rows, h

    def delete_ds(self, ds):
        mask = self.ds.data != ds
        remap = np.where(mask, np.cumsum(mask) - 1, -1).astype(np.int64)
        for a in (self.keys, self.tok, self.ds, self.hash):
            a.keep(mask)
        self.invalidate(True)
        return int((~mask).sum()), remap

    def window(self, w, per):
        return self.tok.data[w * per:(w + 1) * per].reshape(-1)

    def arrays(self):
        return {"keys": self.keys, "tok": self.tok, "ds": self.ds, "hash": self.hash}


class BarameterStore(_KeyedArrays):
    """Hot tier: fixed-capacity slots. key (L2-normalised fp16) + 4x8-bit PQ codes + metadata."""

    def __init__(self, cfg):
        self.cfg, self.cap, self.clock = cfg, cfg.store_slots, 0
        self.keys = GrowArray((cfg.d_k,), np.float16, 4096)
        self.codes = GrowArray((cfg.pq_m,), np.uint8, 4096)
        self.ds = GrowArray((), np.int32, 4096)
        self.usage = GrowArray((), np.int32, 4096)
        self.step = GrowArray((), np.int64, 4096)
        self.chunk = GrowArray((), np.int64, 4096)  # row in ChunkIndex (source text pointer)
        self.hash = GrowArray((), np.int64, 4096)
        self._codes_cache = None

    @property
    def n(self):
        return self.keys.n

    def arrays(self):
        return {"keys": self.keys, "codes": self.codes, "ds": self.ds, "usage": self.usage, "step": self.step,
                "chunk": self.chunk, "hash": self.hash}

    def invalidate(self, hashes_changed=False):
        super().invalidate(hashes_changed)
        self._codes_cache = None

    def _keep(self, mask):
        for a in self.arrays().values():
            a.keep(mask)
        self.invalidate(True)

    def gather_keys(self, idx, device):
        kf = self.keys_for(device)
        if isinstance(kf, torch.Tensor):
            return kf[idx].float()
        flat = np.asarray(kf[idx.cpu().numpy().reshape(-1)])
        return torch.from_numpy(flat).to(device).float().view(*idx.shape, -1)

    def dequant(self, idx, codebook):
        if self._codes_cache is None or len(self._codes_cache) != self.n or self._codes_cache.device != idx.device:
            self._codes_cache = torch.from_numpy(np.ascontiguousarray(self.codes.data)).to(idx.device)
        c = self._codes_cache[idx].long()  # (...,m)
        ar = torch.arange(c.shape[-1], device=idx.device)
        return codebook[ar, c].reshape(*idx.shape, -1)

    def write(self, keys, codes, ds, rows, hashes):
        keys = F.normalize(torch.as_tensor(keys).float(), dim=-1)
        k16 = keys.numpy().astype(np.float16)
        codes = np.asarray(codes, np.uint8)
        self.clock += 1
        slot, hit = self.lookup(hashes)
        for i in np.nonzero(hit)[0]:  # same chunk again: refresh in place
            j = slot[i]
            self.keys.buf[j], self.codes.buf[j] = k16[i], codes[i]
            self.usage.buf[j] += 1
            self.step.buf[j] = self.clock
        todo = np.nonzero(~hit)[0]
        if len(todo) and self.n > 0:  # near-duplicate merge
            sc, ix = self.search(keys[todo], 1)
            sims, ix = sc[:, 0].cpu().numpy(), ix[:, 0].cpu().numpy()
            merge = sims >= self.cfg.dedup_threshold
            for t, j in zip(todo[merge], ix[merge]):
                nk = 0.9 * self.keys.buf[j].astype(np.float32) + 0.1 * keys[t].numpy()
                self.keys.buf[j] = (nk / (np.linalg.norm(nk) + 1e-8)).astype(np.float16)
                self.usage.buf[j] += 1
                self.step.buf[j] = self.clock
            todo = todo[~merge]
        if len(todo):
            todo = todo[-self.cap:]
            need = len(todo) - (self.cap - self.n)
            if need > 0:  # evict least-used, then oldest
                order = np.lexsort((self.step.data, self.usage.data))[:need]
                mask = np.ones(self.n, bool)
                mask[order] = False
                self._keep(mask)
            m = len(todo)
            self.keys.append(k16[todo]); self.codes.append(codes[todo]); self.ds.append(np.full(m, ds, np.int32))
            self.usage.append(np.ones(m, np.int32)); self.step.append(np.full(m, self.clock, np.int64))
            self.chunk.append(rows[todo]); self.hash.append(hashes[todo])
        self.invalidate(True)

    def delete_ds(self, ds):
        mask = self.ds.data != ds
        n = int((~mask).sum())
        if n:
            self._keep(mask)
        return n

    def remap_chunks(self, remap):
        c = self.chunk.data
        ok = c >= 0
        self.chunk.buf[: self.n][ok] = remap[c[ok]]
        gone = self.chunk.data < 0
        if gone.any():
            self._keep(~gone)

    def clear(self):
        self._keep(np.zeros(self.n, bool))


class Memory:
    """Hot store + cold index + dataset tag registry. Lives on the model as model.mem (not an nn.Module)."""

    def __init__(self, cfg):
        self.cfg, self.datasets = cfg, []
        self.store, self.index = BarameterStore(cfg), ChunkIndex(cfg)

    def ds_id(self, tag):
        if tag not in self.datasets:
            self.datasets.append(tag)
        return self.datasets.index(tag)

    def write(self, keys, codes, tokens, tag):
        ds = self.ds_id(tag)
        rows, h = self.index.add(torch.as_tensor(keys).float().numpy(), tokens, ds)
        self.store.write(keys, codes, ds, rows, h)

    def delete(self, tag):
        if tag not in self.datasets:
            return 0, 0
        ds = self.datasets.index(tag)
        ns = self.store.delete_ds(ds)
        ni, remap = self.index.delete_ds(ds)
        self.store.remap_chunks(remap)
        return ns, ni

    def counts(self):
        out = {}
        for i, t in enumerate(self.datasets):
            out[t] = (int((self.store.ds.data == i).sum()), int((self.index.ds.data == i).sum()))
        return out

    def save(self, d):
        os.makedirs(d, exist_ok=True)
        for part, obj in (("store", self.store), ("index", self.index)):
            for name, arr in obj.arrays().items():
                arr.save(os.path.join(d, f"{part}_{name}.npy"))
        json.dump({"datasets": self.datasets, "clock": self.store.clock}, open(os.path.join(d, "memory.json"), "w"))

    def load(self, d, mmap=False):
        if not os.path.exists(os.path.join(d, "memory.json")):
            return
        meta = json.load(open(os.path.join(d, "memory.json")))
        self.datasets, self.store.clock = meta["datasets"], meta["clock"]
        for part, obj in (("store", self.store), ("index", self.index)):
            for name, arr in obj.arrays().items():
                arr.load(os.path.join(d, f"{part}_{name}.npy"), mmap)
            obj.invalidate(True)


# ----------------------------------------------------------------------------------------------
# Barameter read / write modules
# ----------------------------------------------------------------------------------------------
class BarameterWriter(nn.Module):
    """Text chunk -> (key, PQ-coded value). Attention-pool hidden states, project, product-quantise (straight-through)."""

    def __init__(self, cfg):
        super().__init__()
        D, m = cfg.d_model, cfg.pq_m
        self.m, self.K, self.ds_, self.dk, self.dv = m, cfg.pq_k, cfg.d_v // m, cfg.d_k, cfg.d_v
        self.norm = RMSNorm(D)
        self.pool_q = nn.Parameter(torch.randn(D) * 0.02)
        self.wk, self.wv = nn.Linear(D, cfg.d_k, bias=False), nn.Linear(D, cfg.d_v, bias=False)
        self.codebook = nn.Parameter(torch.randn(m, cfg.pq_k, self.ds_) * 0.5)
        self.register_buffer("usage", torch.zeros(m, cfg.pq_k))
        self.last_vs = None

    def forward(self, h, C):
        B, L, D = h.shape
        nc = L // C
        if nc == 0:
            return None
        hh = self.norm(h[:, : nc * C]).float().view(B, nc, C, D)
        a = torch.softmax((hh * self.pool_q.float()).sum(-1) * 0.125, -1)  # fixed temperature: independent of d_model (growth-safe)
        pooled = (a.unsqueeze(-1) * hh).sum(2)
        with torch.autocast(device_type=h.device.type, enabled=False):
            key = F.normalize(self.wk(pooled), dim=-1)
            vs = self.wv(pooled).view(B, nc, self.m, self.ds_)
            cb = self.codebook.float()
            dist = (vs ** 2).sum(-1, keepdim=True) - 2 * torch.einsum("bcmd,mkd->bcmk", vs, cb) + (cb ** 2).sum(-1)
            codes = dist.argmin(-1)  # B,nc,m
            qv = cb[torch.arange(self.m, device=h.device), codes]  # B,nc,m,ds
            commit = 0.25 * F.mse_loss(vs, qv.detach()) + F.mse_loss(qv, vs.detach())
            vq = (vs + (qv - vs).detach()).reshape(B, nc, self.dv)
        div = key.reshape(-1, self.dk).mean(0).pow(2).sum()
        if self.training:
            with torch.no_grad():
                self.usage += F.one_hot(codes, self.K).sum((0, 1)).float()
                self.last_vs = vs.detach().reshape(-1, self.m, self.ds_)[:4096]
        return {"keys": key, "vals": vq, "codes": codes, "aux": {"commit": commit, "div": div}}

    @torch.no_grad()
    def reinit_dead(self):
        if self.last_vs is None:
            return 0
        n = 0
        for mi in range(self.m):
            dead = (self.usage[mi] == 0).nonzero().flatten()
            if len(dead):
                pick = self.last_vs[torch.randint(len(self.last_vs), (len(dead),), device=self.last_vs.device), mi]
                self.codebook.data[mi, dead] = pick + 0.01 * torch.randn_like(pick)
                n += len(dead)
        self.usage.zero_()
        return n


class BarameterRead(nn.Module):
    """x += sigmoid(gate) * Wo( softmax(top-k persistent slots + causal in-sequence chunks) . values ). Starts ~off."""

    def __init__(self, cfg):
        super().__init__()
        D = cfg.d_model
        self.cfg = cfg
        self.norm = RMSNorm(D)
        self.wq = nn.Linear(D, cfg.d_k, bias=False)
        self.gate = nn.Linear(D, 1)
        self.wo = nn.Linear(cfg.d_v, D, bias=False)
        self.log_scale = nn.Parameter(torch.tensor(math.log(10.0)))
        nn.init.constant_(self.gate.bias, -4.0)

    def forward(self, x, mem, codebook, local, use_store):
        cfg = self.cfg
        B, L, _ = x.shape
        h = self.norm(x)
        q = F.normalize(self.wq(h).float(), dim=-1)
        scale = self.log_scale.exp().clamp(max=64.0)
        parts, sp, vp = [], None, None
        if use_store and mem.store.n > 0:
            qf = q.reshape(B * L, -1)
            with torch.no_grad():
                _, idx = mem.store.search(qf.detach(), cfg.topk)
            kk = mem.store.gather_keys(idx, x.device)
            sp = (scale * (qf.unsqueeze(1) * kk).sum(-1)).view(B, L, -1)
            vp = mem.store.dequant(idx, codebook).float().view(B, L, idx.shape[1], -1)
            parts.append(sp)
        if local is not None:
            lk, C = local["keys"], cfg.write_chunk
            nc = lk.shape[1]
            sl = scale * torch.einsum("bld,bcd->blc", q, lk)
            valid = (torch.arange(L, device=x.device) // C)[:, None] > torch.arange(nc, device=x.device)[None]
            parts.append(sl.masked_fill(~valid[None], -1e4))
        if not parts:
            return x, q
        allsc = torch.cat(parts, -1)
        w = torch.softmax(allsc, -1) * (allsc > -1e3)
        r, off = 0, 0
        if sp is not None:
            kp = sp.shape[-1]
            r = r + torch.einsum("blk,blkd->bld", w[..., :kp], vp)
            off = kp
        if local is not None:
            r = r + torch.einsum("blc,bcd->bld", w[..., off:], local["vals"].float())
        out = torch.sigmoid(self.gate(h).float()) * self.wo(r.to(h.dtype)).float()
        return x + out.to(x.dtype), q


# ----------------------------------------------------------------------------------------------
# Model
# ----------------------------------------------------------------------------------------------
class BarameterForCausalLM(PreTrainedModel):
    config_class = BarameterConfig
    base_model_prefix = "barameter"
    _tied_weights_keys = []

    def __init__(self, config):
        super().__init__(config)
        self.all_tied_weights_keys = {}
        cfg = config
        self.embed = nn.Embedding(cfg.vocab_size, cfg.d_model)
        self.blocks = nn.ModuleList([Block(cfg, t) for t in cfg.block_types])
        self.reads = nn.ModuleDict({str(n): BarameterRead(cfg) for n in cfg.read_after})
        self.writer = BarameterWriter(cfg)
        self.norm = RMSNorm(cfg.d_model)
        self.mem = Memory(cfg)
        self.last_writes, self.last_aux = None, {}
        self._init()

    def _init(self):
        std, out_std = 0.02, 0.02 / math.sqrt(2 * self.config.n_layers)
        for name, m in self.named_modules():
            if isinstance(m, nn.Linear):
                tail = name.split(".")[-1]
                nn.init.normal_(m.weight, 0, out_std if tail in ("out", "w2", "wo") else std)
                if m.bias is not None:
                    nn.init.zeros_(m.bias)
            elif isinstance(m, nn.Embedding):
                nn.init.normal_(m.weight, 0, std)
        for b in self.blocks:
            if b.kind == "lin":
                with torch.no_grad():
                    b.mix.g.bias.copy_(torch.linspace(-4.5, -1.0, b.mix.H))
        for n, r in self.reads.items():
            nn.init.constant_(r.gate.bias, -4.0)

    def get_input_embeddings(self):
        return self.embed

    def num_params(self):
        return sum(p.numel() for p in self.parameters())

    # ---- core ----
    def _run(self, ids, use_store=True, C=None, stop_at_write=False):
        cfg = self.config
        C = C or cfg.write_chunk
        x, writes, aux = self.embed(ids), None, {}
        for i, blk in enumerate(self.blocks):
            if cfg.grad_checkpoint and self.training and torch.is_grad_enabled():
                x = checkpoint(blk, x, use_reentrant=False)
            else:
                x = blk(x)
            n = i + 1
            if n == cfg.write_from:
                writes = self.writer(x, C)
                if writes is not None:
                    aux.update(writes["aux"])
                if stop_at_write:
                    return x, writes, aux
            if str(n) in self.reads:
                local = writes if (writes is not None and cfg.local_memory and n > cfg.write_from) else None
                rd = self.reads[str(n)]
                x, q = rd(x, self.mem, self.writer.codebook, local, use_store)
                if local is not None and self.training and local["keys"].shape[1] > 1:
                    nc, B = local["keys"].shape[1], ids.shape[0]
                    qm = F.normalize(q[:, : nc * C].reshape(B, nc, C, -1).mean(2), dim=-1)[:, 1:].reshape(B * (nc - 1), -1)
                    kt = local["keys"][:, :-1].reshape(B * (nc - 1), -1)
                    lg = rd.log_scale.exp().clamp(max=64.0) * qm @ kt.T
                    aux["contrast"] = F.cross_entropy(lg, torch.arange(len(lg), device=lg.device))
        return x, writes, aux

    def forward(self, input_ids, labels=None, use_store=True, reduction="mean", **kw):
        cfg = self.config
        x, writes, aux = self._run(input_ids, use_store)
        h = self.norm(x)
        logits = cfg.softcap * torch.tanh(F.linear(h, self.embed.weight.to(h.dtype)).float() / cfg.softcap)
        loss = None
        if labels is not None:
            lg, tg = logits[:, :-1], labels[:, 1:]
            ce = F.cross_entropy(lg.reshape(-1, lg.shape[-1]), tg.reshape(-1), ignore_index=-100, reduction="none").view(tg.shape)
            valid = (tg != -100).float()
            if reduction == "none":
                loss = (ce * valid).sum(1) / valid.sum(1).clamp(min=1)
            else:
                ce_m = (ce * valid).sum() / valid.sum().clamp(min=1)
                z = torch.logsumexp(lg, -1).pow(2).mean()
                loss = ce_m + 1e-4 * z + aux.get("commit", 0) + 0.1 * aux.get("contrast", 0) + 0.01 * aux.get("div", 0)
                self.last_aux = {"ce": ce_m.detach(), **{k: v.detach() for k, v in aux.items()}}
        self.last_writes = writes
        return CausalLMOutput(loss=loss, logits=logits)

    @torch.no_grad()
    def encode_chunks(self, ids, chunk=None):
        """Forward-only barameter writer (ingestion path). Returns dict(keys, codes) or None."""
        was = self.training
        self.eval()
        _, writes, _ = self._run(ids, True, chunk, stop_at_write=True)
        self.train(was)
        return writes

    @torch.no_grad()
    def chat_generate(self, ids, max_new=128, temperature=0.7, top_k=40, eos=1):
        self.eval()
        out = ids.clone()
        for _ in range(max_new):
            lg = self(out[:, -self.config.max_seq_len:]).logits[:, -1].float()
            if temperature <= 0:
                nxt = lg.argmax(-1, keepdim=True)
            else:
                lg = lg / temperature
                v, _ = lg.topk(min(top_k, lg.shape[-1]))
                lg = lg.masked_fill(lg < v[:, -1:], float("-inf"))
                nxt = torch.multinomial(torch.softmax(lg, -1), 1)
            out = torch.cat([out, nxt], 1)
            if int(nxt) == eos:
                break
        return out

    def load_memory(self, path_or_repo, mmap=False):
        d = resolve_path(path_or_repo)
        self.mem.load(os.path.join(d, "memory"), mmap)


try:
    AutoConfig.register("barameter", BarameterConfig)
except Exception:
    pass
try:
    AutoModelForCausalLM.register(BarameterConfig, BarameterForCausalLM)
except Exception:
    pass


# ----------------------------------------------------------------------------------------------
# Function-preserving growth (depth and width) - no pretraining restart
# ----------------------------------------------------------------------------------------------
ZERO_SUFFIX = ("mix.out.weight", "ffn.w2.weight", "wo.weight")


def grow_model(model, add_layers=0, new_d_model=None):
    old = model.config
    cfg = copy.deepcopy(old)
    D0, dh, g = old.d_model, old.head_dim, old.n_heads // old.n_kv_heads
    if new_d_model and new_d_model > D0:
        H1 = math.ceil(new_d_model / dh / g) * g  # keeps GQA group size so old heads keep their KV mapping
        cfg.d_model, cfg.n_heads, cfg.n_kv_heads = H1 * dh, H1, H1 // g
        cfg.ffn_hidden = int(round(old.ffn_hidden * cfg.d_model / D0 / 64)) * 64
    D1 = cfg.d_model
    n_old = old.n_layers
    ins = sorted({min(n_old - 1, max(1, round((j + 1) * n_old / (add_layers + 1)))) for j in range(add_layers)}) if add_layers else []
    types, new2old, oi = [], {}, 0
    for pos in range(n_old):
        while ins and ins[0] == pos:
            types.append("lin"); ins.pop(0)
        new2old[len(types)] = pos
        types.append(old.block_types[pos])
    while ins:
        types.append("lin"); ins.pop(0)
    inserted_before = lambda c: sum(1 for ni in range(len(types)) if ni not in new2old and ni < _new_index_of_old(new2old, c))
    old2new = {o: n for n, o in new2old.items()}
    shift = lambda c: (old2new[c - 1] + 1) if c >= 1 else 0  # count after old block c-1
    cfg.block_types, cfg.n_layers = types, len(types)
    cfg.read_after = [shift(c) for c in old.read_after]
    cfg.write_from = shift(old.write_from)
    cfg.growth_history = list(old.growth_history) + [{"add_layers": add_layers, "d_model": [D0, D1], "layers": [n_old, len(types)]}]
    new = BarameterForCausalLM(cfg).to(next(model.parameters()).device)
    read_new2old = {shift(c): c for c in old.read_after}
    osd = model.state_dict()

    def old_name(nm):
        p = nm.split(".")
        if p[0] == "blocks":
            if int(p[1]) not in new2old:
                return None
            p[1] = str(new2old[int(p[1])])
        elif p[0] == "reads":
            p[1] = str(read_new2old[int(p[1])])
        return ".".join(p)

    with torch.no_grad():
        for name, p in new.state_dict().items():
            on = old_name(name)
            if name.endswith(ZERO_SUFFIX):
                p.zero_()
            if on is None:
                continue
            src = osd[on].to(p.dtype)
            if name == "embed.weight":
                p.zero_()
            tail = ".".join(name.split(".")[-2:])
            if tail in ("n1.weight", "n2.weight", "norm.weight") and D1 != D0:
                p.fill_(1.0)
                src = src * math.sqrt(D0 / D1)  # zero-padded dims change RMS; compensate so outputs are identical
            assert all(a >= b for a, b in zip(p.shape, src.shape)), name
            p[tuple(slice(0, s) for s in src.shape)] = src
    new.mem = model.mem
    new.mem.cfg = cfg
    return new


def _new_index_of_old(new2old, c):
    return 0


# ----------------------------------------------------------------------------------------------
# Tokenizer, data
# ----------------------------------------------------------------------------------------------
class Tok:
    pad, eos, user, asst = 0, 1, 2, 3

    def __init__(self, enc):
        self.enc = enc

    @property
    def vocab_size(self):
        return self.enc.get_vocab_size()

    def encode(self, t):
        return self.enc.encode(t).ids

    def decode(self, ids):
        return self.enc.decode([int(i) for i in ids], skip_special_tokens=False)

    def save(self, d):
        os.makedirs(d, exist_ok=True)
        self.enc.save(os.path.join(d, "tokenizer.json"))
        try:
            from transformers import PreTrainedTokenizerFast
            PreTrainedTokenizerFast(tokenizer_object=self.enc, pad_token="<pad>", eos_token="<eos>").save_pretrained(d)
        except Exception:
            pass

    @staticmethod
    def load(d):
        from tokenizers import Tokenizer
        return Tok(Tokenizer.from_file(os.path.join(d, "tokenizer.json")))

    @staticmethod
    def train(texts, vocab_size):
        from tokenizers import Tokenizer, models, pre_tokenizers, decoders, trainers
        enc = Tokenizer(models.BPE())
        enc.pre_tokenizer = pre_tokenizers.ByteLevel(add_prefix_space=False)
        enc.decoder = decoders.ByteLevel()
        tr = trainers.BpeTrainer(vocab_size=vocab_size, special_tokens=SPECIALS, show_progress=False,
                                 initial_alphabet=pre_tokenizers.ByteLevel.alphabet())
        enc.train_from_iterator(texts, tr)
        return Tok(enc)


def format_row(r):
    if "instruction" in r:
        u = r["instruction"] + (("\n" + r["input"]) if r.get("input") else "")
        a = r.get("output") or r.get("response")
    elif "question" in r:
        u, a = r["question"], r.get("response") or r.get("answer")
    elif "prompt" in r:
        u, a = r["prompt"], r.get("response") or r.get("completion") or r.get("chosen")
    elif "text" in r:
        return r["text"]
    else:
        return None
    if not u or not a:
        return None
    return f"<|user|>\n{u}\n<|assistant|>\n{a}"


def load_rows(name, split="train", streaming=False, max_examples=0, holdout=500):
    """Returns (make_rows(epoch) -> iterator of strings, heldout list of strings)."""
    from datasets import load_dataset
    ds = load_dataset(name, split=split, streaming=streaming)
    if streaming:
        held = [t for t in (format_row(r) for r in ds.take(holdout)) if t]
        tr = ds.skip(holdout)
        if max_examples:
            tr = tr.take(max_examples)
    else:
        n = len(ds)
        held = [t for t in (format_row(ds[i]) for i in range(min(holdout, n))) if t]
        tr = ds.select(range(min(holdout, n), n if not max_examples else min(n, holdout + max_examples)))

    def make_rows(epoch):
        src = tr.shuffle(seed=epoch, buffer_size=10000) if streaming else tr.shuffle(seed=epoch)
        for r in src:
            t = format_row(r)
            if t:
                yield t

    return make_rows, held


def _pack(tok, texts, L):
    buf, batch = [], []

    def flush():
        for e in tok.enc.encode_batch(batch):
            buf.extend(e.ids)
            buf.append(tok.eos)
        batch.clear()

    for t in texts:
        batch.append(t)
        if len(batch) >= 256:
            flush()
            while len(buf) >= L:
                yield np.asarray(buf[:L], np.int64)
                del buf[:L]
    if batch:
        flush()
    while len(buf) >= L:
        yield np.asarray(buf[:L], np.int64)
        del buf[:L]


def window_stream(tok, make_rows, L, skip=0):
    ep, n = 0, 0
    while True:
        got = False
        for w in _pack(tok, make_rows(ep), L):
            got = True
            n += 1
            if n > skip:
                yield w
        if not got:
            return
        ep += 1


# ----------------------------------------------------------------------------------------------
# Optimiser (Muon for 2-D hidden weights, AdamW for the rest)
# ----------------------------------------------------------------------------------------------
def newton_schulz(G, steps=5):
    a, b, c = 3.4445, -4.7750, 2.0315
    X = G.float()
    X = X / (X.norm() + 1e-7)
    tr = X.shape[0] > X.shape[1]
    if tr:
        X = X.T
    for _ in range(steps):
        A = X @ X.T
        X = a * X + (b * A + c * A @ A) @ X
    return X.T if tr else X


class Muon(torch.optim.Optimizer):
    def __init__(self, params, lr=0.02, momentum=0.95, wd=0.01):
        super().__init__(params, dict(lr=lr, momentum=momentum, wd=wd))

    @torch.no_grad()
    def step(self, closure=None):
        for g in self.param_groups:
            for p in g["params"]:
                if p.grad is None:
                    continue
                st = self.state[p]
                if "buf" not in st:
                    st["buf"] = torch.zeros_like(p)
                st["buf"].mul_(g["momentum"]).add_(p.grad)
                u = newton_schulz(p.grad.add(st["buf"], alpha=g["momentum"]))
                u = u * max(1.0, p.shape[0] / p.shape[1]) ** 0.5
                p.mul_(1 - g["lr"] * g["wd"]).add_(u.to(p.dtype), alpha=-g["lr"])


def build_optimizers(model, lr_muon, lr_adam, wd=0.01):
    mu, ad = [], []
    for n, p in model.named_parameters():
        (mu if p.ndim == 2 and "embed" not in n else ad).append(p)
    return [Muon(mu, lr=lr_muon, wd=wd), torch.optim.AdamW(ad, lr=lr_adam, betas=(0.9, 0.95), weight_decay=0.0)]


def lr_mult(prog, warm=0.03):
    if prog < warm:
        return max(prog / warm, 0.02)
    return 0.1 + 0.9 * 0.5 * (1 + math.cos(math.pi * min(1.0, (prog - warm) / (1 - warm))))


# ----------------------------------------------------------------------------------------------
# Hugging Face Hub (background thread). Auth: env / Colab secret / cached login / interactive login()
# ----------------------------------------------------------------------------------------------
def hf_api():
    from huggingface_hub import HfApi, login
    tok = os.environ.get("HF_TOKEN") or os.environ.get("HUGGINGFACE_HUB_TOKEN")
    if not tok:
        try:
            from google.colab import userdata
            tok = userdata.get("HF_TOKEN")
        except Exception:
            tok = None
    if tok:
        login(token=tok, add_to_git_credential=False)
    api = HfApi()
    try:
        api.whoami()
    except Exception:
        login()
    return api


def resolve_path(path_or_repo, local_dir=None):
    if os.path.isdir(path_or_repo):
        return path_or_repo
    from huggingface_hub import snapshot_download
    hf_api()
    return snapshot_download(repo_id=path_or_repo, local_dir=local_dir or os.path.join("hub_cache", path_or_repo.replace("/", "__")))


class HubSync:
    def __init__(self, repo_id):
        self.repo_id, self.api = repo_id, hf_api()
        self.api.create_repo(repo_id, exist_ok=True, private=True)
        self.q, self.busy = queue.Queue(maxsize=1), threading.Event()
        threading.Thread(target=self._loop, daemon=True).start()

    def _loop(self):
        while True:
            folder, msg = self.q.get()
            self.busy.set()
            try:
                self.api.upload_folder(repo_id=self.repo_id, folder_path=folder, commit_message=msg)
                print(f"[hub] uploaded: {msg}", flush=True)
            except Exception as e:
                print(f"[hub] upload failed ({e}); will retry at the next checkpoint", flush=True)
            finally:
                self.busy.clear()
                self.q.task_done()

    def push(self, folder, msg):
        if self.busy.is_set() or self.q.full():
            return False
        self.q.put_nowait((folder, msg))
        return True

    def flush(self):
        self.q.join()


def save_bundle(d, model, tok):
    from safetensors.torch import save_file
    os.makedirs(d, exist_ok=True)
    save_file({k: v.detach().cpu().contiguous() for k, v in model.state_dict().items()}, os.path.join(d, "model.safetensors"), metadata={"format": "pt"})
    cfg = model.config
    cfg.auto_map = {"AutoConfig": f"{MODULE_NAME}.BarameterConfig", "AutoModelForCausalLM": f"{MODULE_NAME}.BarameterForCausalLM"}
    cfg.save_pretrained(d)
    tok.save(d)
    model.mem.save(os.path.join(d, "memory"))
    src = globals().get("__file__")
    if src and os.path.exists(src):
        shutil.copy(src, os.path.join(d, MODULE_NAME + ".py"))


def load_bundle(path_or_repo, device, mmap=False):
    from safetensors.torch import load_file
    d = resolve_path(path_or_repo)
    cfg = BarameterConfig.from_pretrained(d)
    model = BarameterForCausalLM(cfg)
    model.load_state_dict(load_file(os.path.join(d, "model.safetensors")), strict=True)
    model.mem.load(os.path.join(d, "memory"), mmap)
    return model.to(device), Tok.load(d), d


# ----------------------------------------------------------------------------------------------
# Training (one continuous loop: gradient steps + barameter writes from the same forward pass)
# ----------------------------------------------------------------------------------------------
def pick_device():
    return torch.device("cuda" if torch.cuda.is_available() else "cpu")


def write_batch(model, ids, tag, writes=None):
    writes = writes or model.last_writes
    if writes is None:
        return
    C = model.config.write_chunk
    nc = writes["keys"].shape[1]
    tokens = ids[:, : nc * C].reshape(-1, C).cpu().numpy()
    model.mem.write(writes["keys"].detach().reshape(-1, model.config.d_k).float().cpu(),
                    writes["codes"].detach().reshape(-1, model.config.pq_m).cpu().numpy(), tokens, tag)


@torch.no_grad()
def evaluate(model, tok, texts, device, L, B=8, max_windows=64):
    model.eval()
    wins = []
    for w in _pack(tok, texts, L):
        wins.append(w)
        if len(wins) >= max_windows:
            break
    if not wins:
        return float("nan"), float("nan")
    res = {True: [], False: []}
    for use in (True, False):
        for i in range(0, len(wins), B):
            ids = torch.from_numpy(np.stack(wins[i:i + B])).to(device)
            res[use].append(model(ids, labels=ids, use_store=use, reduction="none").loss.cpu())
    on, off = torch.cat(res[True]).mean().item(), torch.cat(res[False]).mean().item()
    return on, off


def train(model, tok, make_rows, held, a, device):
    os.makedirs(a.out, exist_ok=True)
    model.to(device).train()
    opts = build_optimizers(model, a.lr_muon, a.lr_adam)
    base_lrs = [[g["lr"] for g in o.param_groups] for o in opts]
    amp = device.type == "cuda"
    scaler = torch.amp.GradScaler("cuda", enabled=amp)
    state = {"step": 0, "windows": 0, "elapsed": 0.0}
    sp = os.path.join(a.out, "trainer_state.pt")
    if a.resume and os.path.exists(sp):
        st = torch.load(sp, map_location="cpu", weights_only=False)
        for o, s in zip(opts, st["opts"]):
            o.load_state_dict(s)
        scaler.load_state_dict(st["scaler"])
        torch.set_rng_state(st["rng"])
        state = st["state"]
        print(f"[resume] step {state['step']} windows {state['windows']} elapsed {state['elapsed']:.0f}s", flush=True)
    hub = HubSync(a.hub_repo) if a.hub_repo else None
    stream = window_stream(tok, make_rows, a.seq_len, skip=state["windows"])
    t_start, last_ck, tok_count, t_log = time.time() - state["elapsed"], time.time(), 0, time.time()
    stage = os.path.join(a.out, "hub_stage")

    def checkpoint_now(final=False):
        state["elapsed"] = time.time() - t_start
        if hub and not final and (hub.busy.is_set() or hub.q.full()):
            return
        save_bundle(stage, model, tok)
        torch.save({"opts": [o.state_dict() for o in opts], "scaler": scaler.state_dict(), "rng": torch.get_rng_state(),
                    "state": dict(state)}, os.path.join(stage, "trainer_state.pt"))
        shutil.copy(os.path.join(stage, "trainer_state.pt"), sp)
        if hub:
            if final:
                hub.flush()
                hub.push(stage, f"final step {state['step']}")
                hub.flush()
            else:
                hub.push(stage, f"checkpoint step {state['step']}")

    while True:
        elapsed = time.time() - t_start
        prog = elapsed / (a.minutes * 60) if a.minutes else state["step"] / max(1, a.max_steps)
        if prog >= 1 or (a.max_steps and state["step"] >= a.max_steps):
            break
        try:
            batch = np.stack([next(stream) for _ in range(a.batch)])
        except StopIteration:
            break
        ids = torch.from_numpy(batch).to(device)
        m = lr_mult(prog)
        for o, bl in zip(opts, base_lrs):
            for g, b in zip(o.param_groups, bl):
                g["lr"] = b * m
        with torch.autocast(device_type=device.type, dtype=torch.float16, enabled=amp):
            out = model(ids, labels=ids)
        scaler.scale(out.loss).backward()
        for o in opts:
            scaler.unscale_(o)
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        for o in opts:
            scaler.step(o)
        scaler.update()
        for o in opts:
            o.zero_grad(set_to_none=True)
        with torch.no_grad():
            write_batch(model, ids, a.tag)
        state["step"] += 1
        state["windows"] += a.batch
        tok_count += ids.numel()
        if state["step"] % a.dead_every == 0:
            dead = model.writer.reinit_dead()
            if dead:
                print(f"[codebook] reinitialised {dead} dead codes", flush=True)
        if state["step"] % a.log_every == 0:
            aux = {k: round(float(v), 4) for k, v in model.last_aux.items()}
            print(f"step {state['step']:5d} loss {float(out.loss.detach()):.4f} {aux} lr x{m:.2f} "
                  f"{tok_count / (time.time() - t_log):.0f} tok/s store {model.mem.store.n} index {model.mem.index.n} "
                  f"elapsed {time.time() - t_start:.0f}s", flush=True)
            tok_count, t_log = 0, time.time()
        if a.ckpt_minutes and time.time() - last_ck > a.ckpt_minutes * 60:
            checkpoint_now()
            last_ck = time.time()
    checkpoint_now(final=True)
    if held:
        on, off = evaluate(model, tok, held, device, a.seq_len)
        print(f"[heldout] loss with barameters {on:.4f} | store disabled {off:.4f}", flush=True)
    model.train()
    return state


# ----------------------------------------------------------------------------------------------
# Search / answer / reindex / forgetting scan / refresh
# ----------------------------------------------------------------------------------------------
@torch.no_grad()
def window_ids(mem, w, per, device):
    return torch.from_numpy(mem.index.window(w, per).astype(np.int64))[None].to(device)


@torch.no_grad()
def reencode_windows(model, windows, device, per, tag_ds=None):
    """Re-encode index windows with the current writer; refreshes index keys and existing store slots."""
    mem, C = model.mem, model.config.write_chunk
    for w in windows:
        ids = window_ids(mem, w, per, device)
        wr = model.encode_chunks(ids)
        if wr is None:
            continue
        rows = np.arange(w * per, w * per + wr["keys"].shape[1])
        k = wr["keys"][0].float().cpu().numpy().astype(np.float16)
        mem.index.keys.buf[rows] = k
        mem.index.invalidate()
        slots = np.nonzero(np.isin(mem.store.chunk.data, rows))[0]
        if len(slots):
            pos = {int(r): i for i, r in enumerate(rows)}
            for s in slots:
                i = pos[int(mem.store.chunk.data[s])]
                mem.store.keys.buf[s] = k[i]
                mem.store.codes.buf[s] = wr["codes"][0, i].cpu().numpy().astype(np.uint8)
    mem.store.invalidate()


@torch.no_grad()
def reindex(model, device, seq_len):
    per = seq_len // model.config.write_chunk
    nw = math.ceil(model.mem.index.n / per)
    reencode_windows(model, range(nw), device, per)
    return nw


@torch.no_grad()
def deep_search(model, tok, text, device, topm=3):
    """Encode the query as one chunk, brute-force search the full ChunkIndex. Returns [(score,row,text)]."""
    mem = model.mem
    ids = torch.tensor([tok.encode(text)], device=device)
    wr = model.encode_chunks(ids, chunk=ids.shape[1])
    if wr is None or mem.index.n == 0:
        return []
    sc, ix = mem.index.search(wr["keys"][0].float(), topm)
    return [(float(s), int(i), tok.decode(mem.index.tok.data[int(i)])) for s, i in zip(sc[0], ix[0])]


@torch.no_grad()
def answer(model, tok, question, device, deep=False, score_thr=0.5, ent_thr=4.0, topm=3, expand=2, max_new=128, temperature=0.7):
    model.eval()
    mem = model.mem
    prompt = tok.encode(f"<|user|>\n{question}\n<|assistant|>\n")
    pid = torch.tensor([prompt], device=device)
    info = {"deep": False, "best_score": None, "entropy": None, "hits": []}
    p = torch.softmax(model(pid).logits[0, -1].float(), -1)
    info["entropy"] = float(-(p * torch.log(p + 1e-9)).sum())
    wr = model.encode_chunks(pid, chunk=pid.shape[1])
    if wr is not None and mem.store.n:
        info["best_score"] = float(mem.store.search(wr["keys"][0].float(), 1)[0][0, 0])
    if deep or info["best_score"] is None or info["best_score"] < score_thr or info["entropy"] > ent_thr:
        hits = deep_search(model, tok, question, device, topm)
        info["deep"], info["hits"] = True, hits
        ctx = []
        for _, row, _ in hits:
            ctx += [int(t) for r in range(row, min(row + 1 + expand, mem.index.n)) for t in mem.index.tok.data[r]]
        prompt = (ctx + prompt)[-(model.config.max_seq_len - max_new):]
        pid = torch.tensor([prompt], device=device)
    out = model.chat_generate(pid, max_new=max_new, temperature=temperature)[0, len(prompt):]
    return tok.decode(out).replace("<eos>", "").strip(), info


@torch.no_grad()
def forgetting_scan(model, device, seq_len, n_windows=128, B=8):
    mem = model.mem
    per = seq_len // model.config.write_chunk
    nw = mem.index.n // per
    if nw == 0:
        return []
    model.eval()
    pick = random.sample(range(nw), min(n_windows, nw))
    res = []
    for i in range(0, len(pick), B):
        ws = pick[i:i + B]
        ids = torch.cat([window_ids(mem, w, per, device) for w in ws])
        off = model(ids, labels=ids, use_store=False, reduction="none").loss.cpu()
        on = model(ids, labels=ids, use_store=True, reduction="none").loss.cpu()
        res += [(float(o), float(n), w) for o, n, w in zip(off, on, ws)]
    return sorted(res, reverse=True)  # worst (store-off loss highest) first


def refresh(model, device, seq_len, steps=20, worst=32, replay=32, lr=3e-5, B=8, scan=128):
    mem, per = model.mem, seq_len // model.config.write_chunk
    scanned = forgetting_scan(model, device, seq_len, scan, B)
    if not scanned:
        return {"scanned": 0}
    bad = [w for _, _, w in scanned[:worst]]
    rest = [w for _, _, w in scanned[worst:]] or bad
    before = sum(s[0] for s in scanned[:worst]) / len(bad)
    opt = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=0.0)
    model.train()
    for _ in range(steps):  # half worst, half replay so repairing one thing does not erase another
        ws = random.sample(bad, min(B // 2, len(bad))) + random.sample(rest, min(B - B // 2, len(rest)))
        ids = torch.cat([window_ids(mem, w, per, device) for w in ws])
        loss = model(ids, labels=ids, use_store=False).loss
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        opt.step(); opt.zero_grad(set_to_none=True)
    model.eval()
    reencode_windows(model, bad, device, per)  # rewrite barameters for the repaired chunks (fixes key drift too)
    with torch.no_grad():
        after = sum(float(model(torch.cat([window_ids(mem, w, per, device) for w in bad[i:i + B]]), labels=torch.cat(
            [window_ids(mem, w, per, device) for w in bad[i:i + B]]), use_store=False, reduction="none").loss.sum())
                    for i in range(0, len(bad), B)) / len(bad)
    return {"scanned": len(scanned), "repaired": len(bad), "loss_before": before, "loss_after": after}


# ----------------------------------------------------------------------------------------------
# Self-test (CPU, no network)
# ----------------------------------------------------------------------------------------------
def selftest():
    torch.manual_seed(0); random.seed(0); np.random.seed(0)
    dev = torch.device("cpu")
    ok = lambda name: print(f"  PASS  {name}", flush=True)
    # 0. default-size parameter count
    big = BarameterForCausalLM(BarameterConfig())
    print(f"  default config params: {big.num_params() / 1e6:.1f}M (vocab 16k)")
    del big
    # 1. recurrence correctness
    q, k = F.normalize(torch.randn(2, 37, 3, 16), dim=-1), F.normalize(torch.randn(2, 37, 3, 16), dim=-1)
    v, g = torch.randn(2, 37, 3, 16), -F.softplus(torch.randn(2, 37, 3))
    err = (chunked_gated_linear_attention(q, k, v, g, 8) - reference_recurrence(q, k, v, g)).abs().max().item()
    assert err < 1e-4, err
    ok(f"chunked recurrence == sequential reference (max err {err:.2e})")
    # data
    caps = {f"Land{i}": f"City{(i * 7) % 50}" for i in range(120)}
    rows = [f"<|user|>\nWhat is the capital of {c}?\n<|assistant|>\nThe capital of {c} is {v}." for c, v in caps.items()]
    held, trainrows = rows[:10], rows[10:]
    tok = Tok.train(rows, 400)
    cfg = BarameterConfig(vocab_size=tok.vocab_size, d_model=128, n_heads=4, n_kv_heads=2, head_dim=32, ffn_hidden=192,
                          block_types=["lin", "lin", "attn", "lin", "lin", "attn"], read_after=[2, 4], write_from=3,
                          rec_chunk=16, write_chunk=8, max_seq_len=64, store_slots=300, d_k=32, d_v=64, pq_m=4, pq_k=16, topk=4)
    model = BarameterForCausalLM(cfg)
    ids = torch.randint(4, cfg.vocab_size, (2, 64))
    model.train()
    out = model(ids, labels=ids)
    out.loss.backward()
    assert torch.isfinite(out.loss) and all(torch.isfinite(p.grad).all() for p in model.parameters() if p.grad is not None)
    ok(f"forward/backward finite (loss {out.loss.item():.3f}, ln V = {math.log(cfg.vocab_size):.3f}, aux {sorted(model.last_aux)})")
    model.zero_grad()
    # 2. training loop with barameter writes
    tmp = tempfile.mkdtemp()
    a = argparse.Namespace(out=tmp, hub_repo=None, minutes=0, max_steps=80, batch=8, seq_len=64, lr_muon=0.02, lr_adam=2e-3,
                           ckpt_minutes=0, tag="capitals", log_every=20, dead_every=40, resume=False)
    make_rows = lambda ep: iter(random.Random(ep).sample(trainrows, len(trainrows)))
    st = train(model, tok, make_rows, held, a, dev)
    assert st["step"] == 80 and model.mem.store.n > 0 and model.mem.index.n > 0 and model.mem.store.n <= cfg.store_slots
    ok(f"train loop: store {model.mem.store.n}/{cfg.store_slots} slots, index {model.mem.index.n} chunks")
    # 3. generate + deep search + answer
    s_, i_ = answer(model, tok, "What is the capital of Land3?", dev, deep=True, max_new=12)
    assert i_["deep"] and len(i_["hits"]) > 0
    ok(f"answer() with deep search ran ({len(i_['hits'])} hits)")
    # 4. forgetting scan + refresh
    r = refresh(model, dev, 64, steps=3, worst=4, scan=8, B=4)
    assert r["scanned"] > 0
    ok(f"refresh ran: {r}")
    nw = reindex(model, dev, 64)
    ok(f"reindex re-encoded {nw} windows")
    # 5. growth is function-preserving
    model.eval()
    base = model(ids).logits
    grown = grow_model(model, add_layers=2, new_d_model=256).eval()
    diff = (grown(ids).logits - base).abs().max().item()
    assert diff < 1e-3, diff
    ok(f"grow {model.num_params() / 1e6:.2f}M -> {grown.num_params() / 1e6:.2f}M params ({cfg.n_layers}->{grown.config.n_layers} layers, "
       f"d {cfg.d_model}->{grown.config.d_model}); max logit diff {diff:.2e}")
    gl = grown.train()(ids, labels=ids)
    gl.loss.backward()
    ok("grown model trains (backward finite)")
    # 6. save / load roundtrip
    save_bundle(tmp + "/ck", grown.eval(), tok)
    m2, t2, _ = load_bundle(tmp + "/ck", dev)
    assert (m2.eval()(ids).logits - grown(ids).logits).abs().max().item() < 1e-5 and m2.mem.store.n == grown.mem.store.n
    ok("save_bundle/load_bundle roundtrip (weights + store + index)")
    try:
        m3 = AutoModelForCausalLM.from_pretrained(tmp + "/ck")
        ok("AutoModelForCausalLM.from_pretrained on the saved folder")
    except Exception as e:
        print(f"  WARN  AutoModelForCausalLM.from_pretrained failed: {e}")
    m4 = AutoModelForCausalLM.from_config(cfg)
    assert isinstance(m4, BarameterForCausalLM)
    ok("AutoClass registration (AutoModelForCausalLM.from_config)")
    # 7. delete a dataset -> forgotten
    n0 = model.mem.store.n
    ns, ni = model.mem.delete("capitals")
    assert model.mem.store.n == 0 and model.mem.index.n == 0 and ns == n0
    ok(f"delete dataset tag: removed {ns} barameters and {ni} chunks")
    print("ALL SELF-TESTS PASSED")


# ----------------------------------------------------------------------------------------------
# CLI
# ----------------------------------------------------------------------------------------------
def main(argv=None):
    ap = argparse.ArgumentParser(description="BarameterLM")
    sub = ap.add_subparsers(dest="cmd", required=True)

    def common(p, ck=True):
        if ck:
            p.add_argument("--ckpt", default="brucoder/BarameterLM")
        p.add_argument("--mmap", action="store_true")
        p.add_argument("--push", default=None, help="repo id to push the updated bundle to")

    sub.add_parser("selftest")
    t = sub.add_parser("train"); common(t, False)
    t.add_argument("--dataset", default="yahma/alpaca-cleaned"); t.add_argument("--split", default="train")
    t.add_argument("--streaming", action="store_true"); t.add_argument("--max_examples", type=int, default=0)
    t.add_argument("--holdout", type=int, default=500); t.add_argument("--tag", default=None)
    t.add_argument("--out", default="barameter_run"); t.add_argument("--hub_repo", default="brucoder/BarameterLM")
    t.add_argument("--minutes", type=float, default=10); t.add_argument("--max_steps", type=int, default=0)
    t.add_argument("--batch", type=int, default=8); t.add_argument("--seq_len", type=int, default=512)
    t.add_argument("--vocab", type=int, default=16384); t.add_argument("--lr_muon", type=float, default=0.02)
    t.add_argument("--lr_adam", type=float, default=1e-3); t.add_argument("--ckpt_minutes", type=float, default=5)
    t.add_argument("--log_every", type=int, default=20); t.add_argument("--dead_every", type=int, default=200)
    t.add_argument("--resume", action="store_true"); t.add_argument("--grad_checkpoint", action="store_true")
    t.add_argument("--seed", type=int, default=0)
    ig = sub.add_parser("ingest"); common(ig)
    ig.add_argument("--dataset", required=True); ig.add_argument("--tag", required=True); ig.add_argument("--split", default="train")
    ig.add_argument("--streaming", action="store_true"); ig.add_argument("--max_examples", type=int, default=0)
    ig.add_argument("--seq_len", type=int, default=512); ig.add_argument("--batch", type=int, default=16)
    for name in ("ask", "search"):
        p = sub.add_parser(name); common(p); p.add_argument("question"); p.add_argument("--deep", action="store_true")
        p.add_argument("--max_new", type=int, default=128); p.add_argument("--temperature", type=float, default=0.7)
    ch = sub.add_parser("chat"); common(ch); ch.add_argument("--deep", action="store_true")
    rf = sub.add_parser("refresh"); common(rf)
    rf.add_argument("--steps", type=int, default=20); rf.add_argument("--worst", type=int, default=32)
    rf.add_argument("--scan", type=int, default=128); rf.add_argument("--seq_len", type=int, default=512)
    rf.add_argument("--lr", type=float, default=3e-5)
    ri = sub.add_parser("reindex"); common(ri); ri.add_argument("--seq_len", type=int, default=512)
    ev = sub.add_parser("eval"); common(ev)
    ev.add_argument("--dataset", default="yahma/alpaca-cleaned"); ev.add_argument("--seq_len", type=int, default=512)
    ev.add_argument("--holdout", type=int, default=500)
    sub.add_parser("list").add_argument("--ckpt", default="brucoder/BarameterLM")
    ins = sub.add_parser("inspect"); ins.add_argument("--ckpt", default="brucoder/BarameterLM"); ins.add_argument("slot", type=int)
    de = sub.add_parser("delete"); common(de); de.add_argument("tag")
    gr = sub.add_parser("grow"); common(gr)
    gr.add_argument("--add_layers", type=int, default=0); gr.add_argument("--d_model", type=int, default=0)
    gr.add_argument("--out", default="barameter_grown")
    args = ap.parse_args(argv)
    dev = pick_device()

    if args.cmd == "selftest":
        return selftest()

    if args.cmd == "train":
        torch.manual_seed(args.seed); random.seed(args.seed)
        args.tag = args.tag or args.dataset.split("/")[-1]
        make_rows, held = load_rows(args.dataset, args.split, args.streaming, args.max_examples, args.holdout)
        resume_dir = os.path.join(args.out, "hub_stage")
        if args.resume and not os.path.exists(os.path.join(args.out, "trainer_state.pt")) and args.hub_repo:
            resolve_path(args.hub_repo, resume_dir)
            shutil.copy(os.path.join(resume_dir, "trainer_state.pt"), os.path.join(args.out, "trainer_state.pt")) if os.path.exists(os.path.join(resume_dir, "trainer_state.pt")) else None
        if args.resume and os.path.exists(os.path.join(resume_dir, "model.safetensors")):
            model, tok, _ = load_bundle(resume_dir, dev)
        else:
            sample = []
            for r in make_rows(0):
                sample.append(r)
                if len(sample) >= 20000:
                    break
            tok = Tok.train(sample, args.vocab)
            cfg = BarameterConfig(vocab_size=tok.vocab_size, max_seq_len=args.seq_len, grad_checkpoint=args.grad_checkpoint)
            model = BarameterForCausalLM(cfg)
        print(f"model: {model.num_params() / 1e6:.1f}M parameters, device {dev}, tag '{args.tag}'", flush=True)
        train(model, tok, make_rows, held, args, dev)
        return

    if args.cmd in ("list", "inspect"):
        model, tok, d = load_bundle(args.ckpt, torch.device("cpu"))
        m = model.mem
        if args.cmd == "list":
            print(f"params {model.num_params() / 1e6:.1f}M | store {m.store.n}/{m.store.cap} slots | index {m.index.n} chunks | growth {model.config.growth_history}")
            for tag, (ns, ni) in m.counts().items():
                print(f"  {tag:30s} barameters {ns:8d}  chunks {ni:8d}")
        else:
            s = args.slot
            c = int(m.store.chunk.data[s])
            print(f"slot {s} dataset={m.datasets[int(m.store.ds.data[s])]} usage={int(m.store.usage.data[s])} "
                  f"step={int(m.store.step.data[s])} codes={m.store.codes.data[s].tolist()} chunk_row={c}")
            if c >= 0:
                print("text:", repr(tok.decode(m.index.tok.data[c])))
        return

    model, tok, d = load_bundle(args.ckpt, dev, getattr(args, "mmap", False))
    model.eval()
    dirty = False
    if args.cmd in ("ask", "chat"):
        qs = [args.question] if args.cmd == "ask" else iter(lambda: input("you> "), None)
        for q in qs:
            ans, info = answer(model, tok, q, dev, deep=args.deep, max_new=getattr(args, "max_new", 128), temperature=getattr(args, "temperature", 0.7))
            print(f"[best barameter score {info['best_score']} | entropy {info['entropy']:.2f} | deep search {'yes' if info['deep'] else 'no'}]")
            print(ans)
    elif args.cmd == "search":
        for s, row, text in deep_search(model, tok, args.question, dev, 5):
            print(f"{s:.3f}  row {row}  {text!r}")
    elif args.cmd == "ingest":
        make_rows, _ = load_rows(args.dataset, args.split, args.streaming, args.max_examples, 0)
        wins, n = [], 0
        for w in window_stream(tok, lambda ep: make_rows(0) if ep == 0 else iter(()), args.seq_len):
            wins.append(w)
            if len(wins) == args.batch:
                ids = torch.from_numpy(np.stack(wins)).to(dev)
                write_batch(model, ids, args.tag, model.encode_chunks(ids))
                n += len(wins); wins = []
                if (n // args.batch) % 20 == 0:
                    print(f"ingested {n} windows; store {model.mem.store.n}, index {model.mem.index.n}", flush=True)
        if wins:
            ids = torch.from_numpy(np.stack(wins)).to(dev)
            write_batch(model, ids, args.tag, model.encode_chunks(ids))
        dirty = True
    elif args.cmd == "refresh":
        print(refresh(model, dev, args.seq_len, args.steps, args.worst, lr=args.lr, scan=args.scan)); dirty = True
    elif args.cmd == "reindex":
        print(f"re-encoded {reindex(model, dev, args.seq_len)} windows"); dirty = True
    elif args.cmd == "eval":
        _, held = load_rows(args.dataset, "train", False, 1, args.holdout)
        on, off = evaluate(model, tok, held, dev, args.seq_len)
        print(f"held-out loss: with barameters {on:.4f} | store disabled {off:.4f}")
    elif args.cmd == "delete":
        print("removed (barameters, chunks):", model.mem.delete(args.tag)); dirty = True
    elif args.cmd == "grow":
        grown = grow_model(model, args.add_layers, args.d_model or None)
        print(f"grown to {grown.num_params() / 1e6:.1f}M parameters, history {grown.config.growth_history}")
        save_bundle(args.out, grown, tok)
        if args.push:
            api = hf_api(); api.create_repo(args.push, exist_ok=True, private=True)
            api.upload_folder(repo_id=args.push, folder_path=args.out)
        return
    if dirty:
        save_bundle(d, model, tok)
        if args.push:
            api = hf_api(); api.create_repo(args.push, exist_ok=True, private=True)
            api.upload_folder(repo_id=args.push, folder_path=d)


if __name__ == "__main__" and "ipykernel" not in sys.modules:
    main()
```
